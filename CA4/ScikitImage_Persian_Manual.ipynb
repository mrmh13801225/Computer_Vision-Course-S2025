<h2 dir="rtl" style="text-align: right;">بارگذاری تصویر</h2>

<p dir="rtl" style="text-align: right;">
در این بخش از کتابخانه <code>scikit-image</code> برای بارگذاری تصویر و از <code>matplotlib</code> برای نمایش آن استفاده می‌کنیم. تابع <code>io.imread</code> تصویر را به‌صورت یک آرایه چندبعدی بارگذاری می‌کند که بسته به نوع تصویر می‌تواند سه‌بعدی (تصویر رنگی RGB) یا دوبعدی (خاکستری) باشد.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، تصویر <code>Lenna.png</code> از مسیر فعلی بارگذاری شده و سپس با استفاده از تابع <code>imshow</code> به صورت تمام‌رنگی نمایش داده می‌شود. همچنین با دستور <code>plt.axis('off')</code> خطوط محوری حذف شده‌اند تا تصویر تمیزتر دیده شود.
</p>

```python
from skimage import io
import matplotlib.pyplot as plt

img = io.imread('./images/Lenna.png')

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.axis('off')
plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">تبدیل تصویر رنگی به تصویر خاکستری</h2>

<p dir="rtl" style="text-align: right;">
در بسیاری از کاربردهای پردازش تصویر، مانند <strong>تشخیص لبه‌ها</strong>، <strong>آستانه‌گذاری</strong> یا <strong>بخش‌بندی</strong>، ابتدا لازم است تصویر رنگی به فرم <strong>خاکستری</strong> تبدیل شود. برای این کار از تابع <code>rgb2gray</code> در <code>skimage.color</code> استفاده می‌کنیم.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، ابتدا تصویر رنگی را از مسیر قبلی بارگذاری کرده و سپس آن را به تصویر خاکستری تبدیل می‌کنیم. هنگام نمایش تصویر خاکستری، باید از پارامتر <code>cmap='gray'</code> استفاده کنیم تا رنگ‌ها درست نمایش داده شوند.
</p>


<p dir="rtl" style="text-align: right;">
 تابع <code>rgb2gray</code> تصویر RGB را به یک آرایه‌ی دوبعدی با مقادیر نرمال‌شده بین ۰ و ۱ تبدیل می‌کند، که شدت روشنایی هر پیکسل را نشان می‌دهد. این تبدیل بر پایه‌ی درک انسان از شدت روشنایی کانال‌های رنگی انجام می‌شود، نه صرفاً میانگین‌گیری ساده.
</p>


```python
from skimage import io, color
import matplotlib.pyplot as plt

original = io.imread('./images/Lenna.png')

gray = color.rgb2gray(original)

plt.figure(figsize=(6, 6))
plt.imshow(gray, cmap='gray')
plt.title("Gray scale image", fontsize=12)
plt.axis('off')
plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">اعمال فیلتر Gaussian Blur روی تصویر خاکستری</h2>

<p dir="rtl" style="text-align: right;">
فیلتر <code>Gaussian Blur</code> یکی از رایج‌ترین فیلترهای صاف‌کننده تصویر است که باعث کاهش نویز و هموار شدن تصویر می‌شود. این فیلتر بر پایهٔ توزیع گاوسی عمل می‌کند و میزان تاری تصویر با پارامتر <code>sigma</code> کنترل می‌شود؛ هرچه مقدار <code>sigma</code> بزرگ‌تر باشد، تصویر مات‌تر می‌شود.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال ابتدا تصویر رنگی را بارگذاری کرده، به تصویر خاکستری تبدیل می‌کنیم و سپس فیلتر Gaussian را با مقدار <code>sigma=2</code> روی تصویر خاکستری اعمال می‌کنیم. در نهایت تصویر مات شده را نمایش می‌دهیم.
</p>



```python
from skimage import io, color, filters
import matplotlib.pyplot as plt

img = io.imread('./images/Lenna.png')


gray_img = color.rgb2gray(img)


blurred_img = filters.gaussian(gray_img, sigma=2)


plt.figure(figsize=(6, 6))
plt.imshow(blurred_img, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">تشخیص لبه با فیلتر Sobel</h2>

<p dir="rtl" style="text-align: right;">
تشخیص لبه یکی از مراحل مهم در پردازش تصویر است که به یافتن نقاطی با تغییرات ناگهانی شدت نور در تصویر کمک می‌کند. فیلتر <code>Sobel</code> یک روش کلاسیک برای تشخیص لبه‌ها است که گرادیان شدت تصویر را در دو جهت محاسبه می‌کند.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، ابتدا تصویر رنگی را بارگذاری کرده و به تصویر خاکستری تبدیل می‌کنیم. سپس با استفاده از فیلتر Sobel، لبه‌های تصویر را استخراج می‌کنیم. در خروجی، تصویر اصلی و تصویر لبه‌ها به صورت کنار هم نمایش داده می‌شوند.
</p>



```python
from skimage import io, color, filters
import matplotlib.pyplot as plt

img = io.imread('./images/Lenna.png')

gray_img = color.rgb2gray(img)

edge_img = filters.sobel(gray_img)

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(img)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(edge_img, cmap='gray')
axes[1].set_title('Edge Detection (Sobel)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">قطعه‌بندی تصویر با آستانه‌گذاری اوتسو</h2>

<p dir="rtl" style="text-align: right;">
قطعه‌بندی تصویر یکی از مراحل مهم پردازش تصویر است که هدف آن جدا کردن بخش‌های مختلف تصویر از هم است. یکی از روش‌های متداول قطعه‌بندی، آستانه‌گذاری است که تصویر را به دو بخش پس‌زمینه و پیش‌زمینه تقسیم می‌کند.
</p>

<p dir="rtl" style="text-align: right;">
روش <strong>اوتسو</strong> (<code>Otsu</code>) یک الگوریتم خودکار برای تعیین مقدار آستانه بهینه است. این روش تلاش می‌کند تا آستانه‌ای را پیدا کند که واریانس داخل هر دو بخش (پیش‌زمینه و پس‌زمینه) را به حداقل برساند و در عین حال واریانس بین دو بخش را به حداکثر برساند. این باعث می‌شود قطعه‌بندی تصویر به بهترین شکل ممکن انجام شود بدون نیاز به تعیین دستی مقدار آستانه.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، ابتدا تصویر رنگی را بارگذاری کرده و به تصویر خاکستری تبدیل می‌کنیم. سپس با استفاده از تابع <code>threshold_otsu</code> مقدار آستانه بهینه محاسبه شده و تصویر به صورت دودویی (باینری) با استفاده از این آستانه ساخته می‌شود. در نهایت، تصویر باینری نمایش داده می‌شود که بخش‌های روشن و تاریک را از هم جدا کرده است.
</p>


```python
from skimage import io, color, filters
import matplotlib.pyplot as plt

img = io.imread('./images/Lenna.png')

gray_img = color.rgb2gray(img)

otsu_thresh = filters.threshold_otsu(gray_img)
binary_img = gray_img > otsu_thresh

plt.figure(figsize=(6, 6))
plt.imshow(binary_img, cmap='gray')
plt.title('Otsu Thresholding', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">قطعه‌بندی تصویر با الگوریتم Watershed</h2>

<p dir="rtl" style="text-align: right;">
الگوریتم <strong>Watershed</strong> یکی از روش‌های قدرتمند قطعه‌بندی تصویر است که بر اساس مفهوم توپوگرافی یا نقشه‌های ارتفاع عمل می‌کند. در این الگوریتم، تصویر به‌عنوان یک سطح توپوگرافی در نظر گرفته می‌شود که مقدار شدت هر پیکسل نشان‌دهنده ارتفاع آن نقطه است. هدف، تقسیم‌بندی تصویر به مناطق مختلف با شبیه‌سازی فرآیند «سیلاب» در این سطح است.
</p>

<p dir="rtl" style="text-align: right;">
مراحل کلی کار به این صورت است که ابتدا تصویر باینری یا آستانه‌گذاری شده را تهیه می‌کنیم. سپس فاصله هر پیکسل تا نزدیک‌ترین پس‌زمینه (فاصله اقلیدسی) محاسبه می‌شود که این فاصله به عنوان یک نقشه ارتفاع استفاده می‌شود. پس از آن با قرار دادن نقاط مارکر (markers) در مناطقی که می‌خواهیم به عنوان «دره‌ها» یا مراکز مناطق در نظر گرفته شوند، الگوریتم Watershed اجرا شده و تصویر به قطعات مستقل تقسیم می‌شود.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، از تصویر سکه‌ها (coins) استفاده شده است که نمونه‌ای شناخته شده برای آزمایش قطعه‌بندی است. مراحل اصلی شامل آستانه‌گذاری تصویر برای تبدیل به باینری، محاسبه فاصله اقلیدسی، تعریف مارکرهای پس‌زمینه و پیش‌زمینه، و در نهایت اجرای الگوریتم Watershed است. خروجی شامل تصویر باینری و تصویر قطعه‌بندی شده رنگی می‌باشد که هر بخش با رنگی مجزا نمایش داده شده است.
</p>


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, morphology, segmentation, color
from scipy import ndimage as ndi

coins = data.coins()

binary_coins = coins > filters.threshold_otsu(coins)

distance_map = ndi.distance_transform_edt(binary_coins)

markers = np.zeros_like(coins, dtype=int)
markers[distance_map < 30] = 1
markers[distance_map > 70] = 2

labels = segmentation.watershed(-distance_map, markers)

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(binary_coins, cmap='gray')
axes[0].set_title('Thresholded Image')
axes[0].axis('off')

axes[1].imshow(color.label2rgb(labels, image=coins, bg_label=0), cmap='nipy_spectral')
axes[1].set_title('Watershed Segmentation')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">تشخیص گوشه‌ها با الگوریتم Harris</h2>

<p dir="rtl" style="text-align: right;">
تشخیص گوشه‌ها یکی از گام‌های مهم در تحلیل تصاویر و بینایی ماشین است که به شناسایی نقاطی از تصویر می‌پردازد که تغییرات شدت روشنایی در چند جهت مختلف به طور همزمان زیاد است. این نقاط معمولاً به عنوان ویژگی‌های محلی مهم برای تطبیق تصویر و تشخیص اشیاء استفاده می‌شوند.
</p>

<p dir="rtl" style="text-align: right;">
الگوریتم <code>Harris</code> با تحلیل ماتریس ساختار (structure tensor) در هر پیکسل، مقادیری را محاسبه می‌کند که بیانگر احتمال وجود گوشه در آن نقطه هستند. سپس با استفاده از تابع <code>corner_peaks</code> نقاط برجسته (گوشه‌ها) استخراج می‌شوند.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، ابتدا تصویر را بارگذاری کرده و به تصویر خاکستری تبدیل می‌کنیم. سپس با استفاده از تابع <code>corner_harris</code> نقشه احتمال گوشه‌ها را محاسبه می‌کنیم. در نهایت، موقعیت دقیق گوشه‌ها استخراج شده و روی تصویر اصلی با نقاط قرمز نشان داده می‌شوند.
</p>



```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color
from skimage.feature import corner_harris, corner_peaks

img = io.imread('./images/Building.png')
gray_img = color.rgb2gray(img)

harris_response = corner_harris(gray_img)

corners = corner_peaks(harris_response, min_distance=5, threshold_rel=0.1)

fig, ax = plt.subplots(figsize=(8, 6))
ax.imshow(gray_img, cmap='gray')
ax.plot(corners[:, 1], corners[:, 0], 'r.', markersize=6)
ax.set_title('Harris Corner Detection')
ax.axis('off')
plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">استخراج ویژگی با HOG</h2>

<p dir="rtl" style="text-align: right;">
HOG یا <strong>هیستوگرام گرادیان‌های جهت‌دار</strong>، یک توصیف‌گر ویژگی قدرتمند برای تحلیل ساختار و شکل اشیاء در تصویر است. این روش در بسیاری از کاربردهای بینایی ماشین مانند <strong>تشخیص اشیاء</strong> (مانند افراد یا چهره‌ها)، <strong>بازشناسی تصاویر</strong> و <strong>ردیابی اجسام</strong> کاربرد دارد.
</p>

<p dir="rtl" style="text-align: right;">
HOG با تحلیل تغییرات شدت روشنایی (گرادیان) در نواحی کوچک تصویر، توزیع جهت‌های گرادیان را به شکل هیستوگرام ذخیره می‌کند. این توزیع‌ها به خوبی ویژگی‌های محلی تصویر را که نشان‌دهنده لبه‌ها و ساختارها هستند، بازنمایی می‌کنند.
</p>

<p dir="rtl" style="text-align: right;">
در این مثال، ابتدا یک تصویر نمونه بارگذاری و خاکستری می‌شود. سپس با تنظیم پارامترهایی مانند <code>orientations</code> (تعداد جهت‌ها)، <code>pixels_per_cell</code> (اندازه سلول)، و <code>cells_per_block</code> (بلوک‌های نرمال‌سازی)، ویژگی‌های HOG محاسبه می‌شود. تصویر خروجی نیز با استفاده از <code>rescale_intensity</code> برای نمایش بهتر بازتنظیم می‌شود.
</p>


```python
import matplotlib.pyplot as plt
from skimage import data, color, exposure, feature

sample_img = io.imread('./images/Lenna.png')
gray_img = color.rgb2gray(sample_img)

hog_features, hog_vis = feature.hog(
    gray_img,
    orientations=8,
    pixels_per_cell=(4, 4),
    cells_per_block=(4, 4),
    block_norm='L2-Hys',
    visualize=True
)

hog_vis_rescaled = exposure.rescale_intensity(hog_vis, in_range=(0, 10))

plt.figure(figsize=(6, 6))
plt.imshow(hog_vis_rescaled, cmap='gray')
plt.title('HOG Feature Visualization')
plt.axis('off')
plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">عملیات مورفولوژیکی: گسترش (Dilation) و فرسایش (Erosion)</h2>

<p dir="rtl" style="text-align: right;">
عملیات‌های مورفولوژیکی در پردازش تصویر معمولاً روی تصاویر دودویی اعمال می‌شوند و ابزارهای قدرتمندی برای تقویت ساختارها یا حذف نویز هستند. در اینجا به دو عملیات پرکاربرد <strong>گسترش (Dilation)</strong> و <strong>فرسایش (Erosion)</strong> می‌پردازیم:
</p>

<ul dir="rtl" style="text-align: right;">
  <li><strong>گسترش (Dilation):</strong> مرزهای اجسام را بزرگ‌تر کرده و باعث پر شدن حفره‌های کوچک یا اتصال اجزای نزدیک به هم می‌شود.</li>
  <li><strong>فرسایش (Erosion):</strong> مرزهای اجسام را کوچک‌تر کرده و معمولاً برای حذف نویزهای ریز یا جداسازی اجسام چسبیده استفاده می‌شود.</li>
</ul>

<p dir="rtl" style="text-align: right;">
در کد زیر، هر دو عملیات بر روی دو تصویر جداگانه اعمال می‌شوند. ابتدا تصویر گسترش یافته و سپس یک تصویر متفاوت برای فرسایش استفاده شده است.
</p>




```python
from skimage import io
from skimage.color import rgb2gray
from skimage.morphology import dilation, erosion, disk
import matplotlib.pyplot as plt

# --- Dilation Section ---
image_dil = io.imread('./images/Lenna.png')
gray_dil = rgb2gray(image_dil)
binary_dil = gray_dil > 0.5
dilated = dilation(binary_dil, disk(3))

# --- Erosion Section ---
image_ero = io.imread('./images/Lenna.png')
gray_ero = rgb2gray(image_ero) if image_ero.ndim == 3 else image_ero
binary_ero = gray_ero > 0.5
eroded = erosion(binary_ero, disk(3))

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

axes[0, 0].imshow(binary_dil, cmap='gray')
axes[0, 0].set_title('Original (for Dilation)')
axes[0, 0].axis('off')

axes[0, 1].imshow(dilated, cmap='gray')
axes[0, 1].set_title('After Dilation')
axes[0, 1].axis('off')

axes[1, 0].imshow(gray_ero, cmap='gray')
axes[1, 0].set_title('Original (for Erosion)')
axes[1, 0].axis('off')

axes[1, 1].imshow(eroded, cmap='gray')
axes[1, 1].set_title('After Erosion')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">تبدیلات هندسی: چرخش و تبدیل آفین</h2>

<p dir="rtl" style="text-align: right;">
در بین عملیات پایه‌ای پردازش تصویر، تبدیلات هندسی نقش مهمی در پیش‌پردازش، تنظیم داده‌ها، یا افزایش داده (Data Augmentation) دارند. دو مورد از این تبدیلات مهم عبارت‌اند از:
</p>

<ul dir="rtl" style="text-align: right;">
  <li><strong>چرخش (Rotation):</strong> تصویر حول یک نقطه (معمولاً مرکز) با زاویه دلخواه دوران داده می‌شود. این عملیات می‌تواند جهت‌گیری اشیاء را تغییر دهد و در کاربردهایی مانند تشخیص الگو یا یادگیری ماشین مفید است.</li>
  <li><strong>تبدیل آفین (Affine Transformation):</strong> نگاشتی خطی است که شامل مقیاس‌دهی، چرخش، انتقال، برش و ترکیبی از این‌ها می‌شود. این تبدیل موقعیت پیکسل‌ها را حفظ می‌کند و اشکال را به صورت خطی تغییر می‌دهد.</li>
</ul>

<p dir="rtl" style="text-align: right;">
در این مثال، ابتدا یک تصویر رنگی با استفاده از تابع <code>rotate</code> چرخانده می‌شود، سپس همان تصویر با تبدیل آفین و مقیاس‌دهی ۱٫۵ برابر روی هر دو محور تغییر داده می‌شود.
</p>

```python
from skimage import data
from skimage.transform import rotate, AffineTransform, warp
import matplotlib.pyplot as plt

image = io.imread('./images/Lenna.png')

rotated = rotate(image, angle=45)

affine = AffineTransform(scale=(1.5, 1.5))
affine_transformed = warp(image, inverse_map=affine)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(image)
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(rotated)
axes[1].set_title('Rotated Image (45°)')
axes[1].axis('off')

axes[2].imshow(affine_transformed)
axes[2].set_title('Affine Transformed (Scaled ×1.5)')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

<h2 dir="rtl" style="text-align: right;">هم‌ترازسازی تصاویر با استفاده از Phase Cross-Correlation</h2>

<p dir="rtl" style="text-align: right;">
در بسیاری از کاربردهای پردازش تصویر، لازم است دو تصویر با یکدیگر هم‌تراز (Aligned) شوند؛ به عنوان مثال در تصویربرداری پزشکی، تصاویر از یک ناحیه بدن در زمان‌های مختلف گرفته می‌شوند و باید با هم مقایسه شوند. <strong>Cross-Correlation فازی</strong> یکی از دقیق‌ترین روش‌ها برای یافتن جابه‌جایی بین دو تصویر است. این روش با استفاده از تحلیل طیفی، شیفت بین تصاویر را با دقت پیکسلی یا حتی زیرپیکسلی محاسبه می‌کند.
</p>

<p dir="rtl" style="text-align: right;">
در این کد:
</p>

<ul dir="rtl" style="text-align: right;">
  <li>ابتدا تصاویر رنگی خوانده شده و به خاکستری تبدیل می‌شوند.</li>
  <li>سپس اندازه دو تصویر برابر می‌شود تا امکان مقایسه فراهم شود.</li>
  <li>با استفاده از تابع <code>phase_cross_correlation</code> شیفت (translation) به‌دست می‌آید.</li>
  <li>در نهایت، تصویر دوم با استفاده از تبدیل آفین (AffineTransform) بر اساس این شیفت جابه‌جا و هم‌تراز می‌شود.</li>
</ul>



```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color
from skimage.transform import warp, AffineTransform, resize
from skimage.registration import phase_cross_correlation

reference = io.imread('./images/reference_image.jpg')
moving = io.imread('./images/shifted_image.jpg')

if reference.ndim == 3:
    reference = color.rgb2gray(reference)
if moving.ndim == 3:
    moving = color.rgb2gray(moving)

if reference.shape != moving.shape:
    moving = resize(moving, reference.shape, anti_aliasing=True)

shift, error, _ = phase_cross_correlation(reference, moving)

tform = AffineTransform(translation=shift)
aligned = warp(moving, tform)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(reference, cmap='gray')
axes[0].set_title('Reference Image')
axes[0].axis('off')

axes[1].imshow(moving, cmap='gray')
axes[1].set_title('Unaligned Image')
axes[1].axis('off')

axes[2].imshow(aligned, cmap='gray')
axes[2].set_title('Aligned Image')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

<h3 dir="rtl"> بازسازی تصویر با حذف بلور (Deconvolution)</h3>

<p dir="rtl">
در این مثال، از <strong>Deconvolution</strong> یا بازسازی تصویر استفاده می‌کنیم تا یک تصویر بلور شده را به وضعیت اولیه‌اش بازگردانیم.
ابتدا تصویر را با فیلتر گوسی (Gaussian blur) محو می‌کنیم، سپس با استفاده از <strong>فیلتر Wiener</strong> تلاش می‌کنیم تصویر اصلی را بازیابی کنیم.
</p>

<p dir="rtl">
این تکنیک زمانی مفید است که تصویر در اثر عواملی مانند حرکت، تار شدن نوری یا لنزهای ضعیف، مات شده باشد.
در اینجا برای شبیه‌سازی فرایند تاری، از یک تابع پخش نقطه‌ای ساده (Point Spread Function) استفاده می‌کنیم.
</p>



```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
from skimage.color import rgb2gray
from skimage.filters import gaussian
from skimage.restoration import wiener

image = io.imread("./images/Lenna.png")

if image.ndim == 3:
    image = rgb2gray(image)

blurred = gaussian(image, sigma=3)

psf = np.ones((5, 5)) / 25

restored = wiener(blurred, psf, balance=0.1)
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(image, cmap='gray')
axes[0].set_title('Original Image')
axes[0].axis('off')

axes[1].imshow(blurred, cmap='gray')
axes[1].set_title('Blurred Image')
axes[1].axis('off')

axes[2].imshow(restored, cmap='gray')
axes[2].set_title('Deconvolved Image')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

<h2 dir="rtl"> تشخیص لکه‌های پوستی مشکوک</h2>

<p dir="rtl">
در این پروژه، هدف شناسایی نواحی مشکوک پوست (مثلاً خال یا لکه‌هایی با شدت رنگ متفاوت) با استفاده از ابزارهای پایه‌ای پردازش تصویر است. تصویر ورودی ابتدا به خاکستری تبدیل شده و با فیلتر گوسی (Gaussian) نرم می‌شود تا نویز کاهش یابد. سپس با روش <code>Otsu Thresholding</code> نواحی مشکوک جدا شده و نواحی کوچکتر از حد خاصی حذف می‌شوند تا فقط نواحی مهم باقی بمانند.
</p>

<p dir="rtl">
در مرحله‌ی نهایی، از <code>labeling</code> برای تشخیص نواحی استفاده شده و آن‌ها با رنگ روی تصویر اصلی برجسته می‌شوند.
</p>




```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, color, filters, measure, morphology
from skimage.filters import threshold_otsu, gaussian

image = data.skin()

gray_image = color.rgb2gray(image)

blurred = gaussian(gray_image, sigma=1)

threshold = threshold_otsu(blurred)
lesion_mask = blurred < threshold

cleaned_mask = morphology.remove_small_objects(lesion_mask, min_size=500, connectivity=2)

labeled_regions = measure.label(cleaned_mask)
overlay = color.label2rgb(labeled_regions, image=image, bg_label=0)

fig, ax = plt.subplots(1, 2, figsize=(14, 6))
ax[0].imshow(gray_image, cmap='gray')
ax[0].set_title('Grayscale Image')
ax[0].axis('off')

ax[1].imshow(overlay)
ax[1].set_title('Detected Lesions Overlay')
ax[1].axis('off')

plt.tight_layout()
plt.show()

regions = measure.regionprops(labeled_regions)
print(f"Number of detected suspicious regions: {len(regions)}")
```

<h2 dir="rtl"> شمارش اشیاء (مثال: سکه‌ها)</h2>

<p dir="rtl">
در این پروژه، هدف شمارش خودکار اشیاء جدا از هم در یک تصویر است؛ مثل سکه‌ها یا قرص‌هایی که روی پس‌زمینه ساده قرار دارند.
با ترکیب چند ابزار در کتابخانه <code>scikit-image</code> مراحل زیر را انجام می‌دهیم:
</p>

<ol dir="rtl">
  <li>تصویر خاکستری شده و با فیلتر Gaussian نرم می‌شود تا نویز حذف شود.</li>
  <li>سپس با روش <code>threshold_otsu</code> به تصویر باینری تبدیل می‌شود.</li>
  <li>با استفاده از <code>distance transform</code> فاصله از پس‌زمینه اندازه‌گیری می‌شود.</li>
  <li>الگوریتم <code>watershed</code> برای جدا کردن اجسام چسبیده استفاده می‌شود.</li>
  <li>در نهایت، با برچسب‌گذاری نواحی، تعداد اجسام محاسبه می‌شود.</li>
</ol>



```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, morphology, segmentation, measure, color
from scipy import ndimage as ndi

image = data.coins()
blurred = filters.gaussian(image, sigma=1)
thresh = filters.threshold_otsu(blurred)
binary = blurred > thresh
distance = ndi.distance_transform_edt(binary)
markers = morphology.label(distance > 6)
labels = segmentation.watershed(-distance, markers, mask=binary)
object_count = np.max(labels)
print(f"Number of detected objects: {object_count}")

colored = color.label2rgb(labels, image=image, bg_label=0)

fig, ax = plt.subplots(1, 2, figsize=(14, 6))
ax[0].imshow(image, cmap='gray')
ax[0].set_title('Original Image')
ax[0].axis('off')

ax[1].imshow(colored)
ax[1].set_title('Labeled Objects')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

<h2 dir="rtl"> تشخیص بیماری یا لکه در برگ</h2>

<p dir="rtl">
در این پروژه، هدف تشخیص نواحی مشکوک یا آلوده در برگ گیاه با استفاده از روش‌های پردازش تصویر کلاسیک است. تصویر برگ با لکه‌های قهوه‌ای یا زرد را تحلیل می‌کنیم تا نواحی غیرمعمول شناسایی شوند و در قالب یک نقشه لکه (heatmap) نمایش داده شوند.
</p>

<ol dir="rtl">
  <li>بارگذاری تصویر رنگی برگ با نشانه‌های بیماری</li>
  <li>تبدیل به فضای HSV یا LAB برای استخراج کانال رنگ</li>
  <li>آستانه‌گذاری adaptive یا Otsu روی کانال رنگ (مثلاً کانال V یا B)</li>
  <li>تولید ماسک نواحی مشکوک و حذف نویزهای کوچک</li>
  <li>برچسب‌گذاری هر لکه و اندازه‌گیری ویژگی‌ها مانند مساحت و موقعیت</li>
  <li>ساخت یک heatmap: شدت لکه بر حسب میزان تغییر رنگ یا مساحت ناحیه</li>
</ol>


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, filters, measure, morphology, exposure
from skimage.color import rgb2gray
from skimage.filters import threshold_otsu
from skimage.morphology import remove_small_objects, disk
from scipy import ndimage as ndi

leaf = io.imread('./images/leaf.jpeg')

lab = color.rgb2lab(leaf)
b_channel = lab[:, :, 2]

blur = filters.gaussian(b_channel, sigma=1)
thresh = threshold_otsu(blur)
mask = blur > thresh

mask_clean = remove_small_objects(mask, min_size=5)

labels = measure.label(mask_clean)
props = measure.regionprops(labels)

heatmap = np.zeros_like(mask_clean, dtype=float)
for region in props:
    heatmap[labels == region.label] = region.area

heatmap_norm = exposure.rescale_intensity(heatmap, out_range=(0, 1))

lesion_count = len(props)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(leaf)
axes[0].set_title('Original Leaf')
axes[0].axis('off')

axes[1].imshow(mask_clean, cmap='gray')
axes[1].set_title('Detected Lesions Mask')
axes[1].axis('off')

axes[2].imshow(leaf)
axes[2].imshow(heatmap_norm, cmap='hot', alpha=0.6)
axes[2].set_title('Heatmap Overlay')
axes[2].axis('off')

plt.tight_layout()
plt.show()

print(f"Number of detected lesions: {lesion_count}")
```
