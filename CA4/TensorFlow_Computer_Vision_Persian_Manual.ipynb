### ۲.۱ مفهوم Tensor و انواع آن

**Tensor** واحد اصلی داده در TensorFlow است. یک Tensor آرایه چندبعدی است که می‌تواند:
- **اسکالر** (0 بعد): یک عدد ساده
- **بردار** (1 بعد): آرایه یک‌بعدی
- **ماتریس** (2 بعد): آرایه دوبعدی
- **Tensor مرتبه بالا** (3+ بعد): آرایه‌های چندبعدی

برای بینایی ماشین، تصاویر معمولاً به صورت Tensor 3 یا 4 بعدی نمایش داده می‌شوند:
- **تصویر خاکستری**: `(height, width)`
- **تصویر رنگی**: `(height, width, channels)`
- **دسته از تصاویر**: `(batch_size, height, width, channels)`

```python
# ۲.۱.۱ ایجاد انواع مختلف Tensor
import tensorflow as tf
import numpy as np

# اسکالر (0 بعد)
scalar = tf.constant(42.0)
print("اسکالر:", scalar)
print("شکل:", scalar.shape)
print("نوع داده:", scalar.dtype)
print("تعداد ابعاد:", scalar.ndim)
print()

# بردار (1 بعد)
vector = tf.constant([1, 2, 3, 4, 5])
print("بردار:", vector)
print("شکل:", vector.shape)
print()

# ماتریس (2 بعد)
matrix = tf.constant([[1, 2, 3], [4, 5, 6]])
print("ماتریس:")
print(matrix)
print("شکل:", matrix.shape)
print()

# Tensor 3 بعدی (مناسب برای یک تصویر رنگی)
image_tensor = tf.random.uniform((224, 224, 3))  # height, width, channels
print("تصویر 3 بعدی - شکل:", image_tensor.shape)
print("حداقل مقدار:", tf.reduce_min(image_tensor))
print("حداکثر مقدار:", tf.reduce_max(image_tensor))
print()

# Tensor 4 بعدی (دسته از تصاویر)
batch_images = tf.random.uniform((32, 224, 224, 3))  # batch_size, height, width, channels
print("دسته تصاویر - شکل:", batch_images.shape)
print("حجم حافظه (تقریبی):", batch_images.numpy().nbytes / (1024**2), "مگابایت")
```

# بخش ۵: آموزش و ارزیابی جامع مدل‌ها

در این بخش به طور تفصیلی با فرآیند آموزش مدل‌های بینایی ماشین، تکنیک‌های بهینه‌سازی، و روش‌های ارزیابی دقیق آشنا می‌شوید.

## ۵.۱. مفاهیم کلیدی آموزش مدل

### فرآیند آموزش:
- **Forward Pass**: محاسبه پیش‌بینی‌ها
- **Loss Calculation**: محاسبه خطا
- **Backward Pass**: محاسبه گرادیان‌ها
- **Parameter Update**: به‌روزرسانی وزن‌ها

### انواع یادگیری:
- **Supervised Learning**: آموزش با برچسب
- **Semi-supervised**: ترکیب داده‌های برچسب‌دار و بدون برچسب
- **Transfer Learning**: استفاده از مدل‌های پیش‌آموزش‌دیده
- **Fine-tuning**: تنظیم دقیق مدل‌های موجود

### معیارهای ارزیابی:
- **Accuracy**: درصد پیش‌بینی‌های صحیح
- **Precision**: دقت در کلاس مثبت
- **Recall**: حساسیت تشخیص
- **F1-Score**: میانگین هارمونیک Precision و Recall
- **AUC-ROC**: منطقه زیر منحنی ROC

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, optimizers, metrics
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from typing import Dict, List, Tuple, Optional
import time
import os
from datetime import datetime

class ComprehensiveTrainer:
    """
    کلاس جامع برای آموزش و ارزیابی مدل‌های بینایی ماشین
    """
    
    def __init__(self, model: keras.Model, model_name: str = "model"):
        """
        مقداردهی اولیه trainer
        
        Args:
            model: مدل Keras برای آموزش
            model_name: نام مدل برای ذخیره‌سازی
        """
        self.model = model
        self.model_name = model_name
        self.history = None
        self.training_time = 0
        self.best_weights = None
        
        # تنظیم مسیر ذخیره
        self.save_dir = f"model_outputs/{model_name}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
        os.makedirs(self.save_dir, exist_ok=True)
    
    def prepare_callbacks(self, 
                         monitor: str = 'val_accuracy',
                         patience: int = 10,
                         reduce_lr_patience: int = 5,
                         min_lr: float = 1e-7) -> List[callbacks.Callback]:
        """
        آماده‌سازی callback های مفید برای آموزش
        
        Args:
            monitor: معیار نظارت
            patience: صبر برای early stopping
            reduce_lr_patience: صبر برای کاهش learning rate
            min_lr: حداقل learning rate
            
        Returns:
            لیست callback ها
        """
        callback_list = [
            # Early Stopping برای جلوگیری از overfitting
            callbacks.EarlyStopping(
                monitor=monitor,
                patience=patience,
                restore_best_weights=True,
                verbose=1
            ),
            
            # کاهش learning rate هنگام توقف پیشرفت
            callbacks.ReduceLROnPlateau(
                monitor=monitor,
                factor=0.5,
                patience=reduce_lr_patience,
                min_lr=min_lr,
                verbose=1
            ),
            
            # ذخیره بهترین مدل
            callbacks.ModelCheckpoint(
                filepath=os.path.join(self.save_dir, 'best_model.h5'),
                monitor=monitor,
                save_best_only=True,
                save_weights_only=False,
                verbose=1
            ),
            
            # CSV Logger برای ذخیره تاریخچه
            callbacks.CSVLogger(
                filename=os.path.join(self.save_dir, 'training_log.csv'),
                append=True
            )
        ]
        
        return callback_list
    
    def compile_model(self, 
                     optimizer: str = 'adam',
                     learning_rate: float = 0.001,
                     loss: str = 'sparse_categorical_crossentropy',
                     metrics: List[str] = ['accuracy']):
        """
        کامپایل مدل با تنظیمات پیشرفته
        
        Args:
            optimizer: نوع بهینه‌ساز
            learning_rate: نرخ یادگیری
            loss: تابع هزینه
            metrics: معیارهای ارزیابی
        """
        # انتخاب optimizer
        if optimizer == 'adam':
            opt = optimizers.Adam(learning_rate=learning_rate)
        elif optimizer == 'sgd':
            opt = optimizers.SGD(learning_rate=learning_rate, momentum=0.9)
        elif optimizer == 'rmsprop':
            opt = optimizers.RMSprop(learning_rate=learning_rate)
        elif optimizer == 'adamw':
            opt = optimizers.experimental.AdamW(learning_rate=learning_rate, weight_decay=0.01)
        else:
            opt = optimizer
        
        # اضافه کردن معیارهای اضافی
        extended_metrics = metrics.copy()
        if 'precision' not in extended_metrics:
            extended_metrics.append(metrics.Precision(name='precision'))
        if 'recall' not in extended_metrics:
            extended_metrics.append(metrics.Recall(name='recall'))
        
        self.model.compile(
            optimizer=opt,
            loss=loss,
            metrics=extended_metrics
        )
        
        print(f"✅ مدل با optimizer {optimizer} و learning rate {learning_rate} کامپایل شد")
    
    def train_model(self,
                   train_dataset,
                   validation_dataset,
                   epochs: int = 100,
                   verbose: int = 1,
                   use_callbacks: bool = True) -> keras.callbacks.History:
        """
        آموزش مدل با تنظیمات پیشرفته
        
        Args:
            train_dataset: دیتاست آموزش
            validation_dataset: دیتاست اعتبارسنجی
            epochs: تعداد epoch ها
            verbose: سطح جزئیات نمایش
            use_callbacks: استفاده از callback ها
            
        Returns:
            تاریخچه آموزش
        """
        print(f"🚀 شروع آموزش مدل {self.model_name}")
        print(f"📊 تعداد epoch ها: {epochs}")
        
        # آماده‌سازی callback ها
        callback_list = self.prepare_callbacks() if use_callbacks else []
        
        # ثبت زمان شروع
        start_time = time.time()
        
        # آموزش مدل
        self.history = self.model.fit(
            train_dataset,
            validation_data=validation_dataset,
            epochs=epochs,
            callbacks=callback_list,
            verbose=verbose
        )
        
        # محاسبه زمان آموزش
        self.training_time = time.time() - start_time
        
        print(f"\n✅ آموزش تکمیل شد!")
        print(f"⏱️ زمان آموزش: {self.training_time:.2f} ثانیه")
        print(f"📈 بهترین دقت اعتبارسنجی: {max(self.history.history['val_accuracy']):.4f}")
        
        return self.history

# مثال استفاده از ComprehensiveTrainer
print("=== آماده‌سازی Trainer جامع ===")

# فرض: مدل CNN از بخش قبل
if 'model_cnn' in locals():
    trainer = ComprehensiveTrainer(model_cnn, "CNN_Enhanced")
else:
    # ایجاد مدل نمونه
    model_sample = keras.Sequential([
        layers.Conv2D(32, 3, activation='relu', input_shape=(224, 224, 3)),
        layers.MaxPooling2D(),
        layers.Conv2D(64, 3, activation='relu'),
        layers.MaxPooling2D(),
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    trainer = ComprehensiveTrainer(model_sample, "Sample_CNN")

print(f"✅ Trainer برای مدل {trainer.model_name} آماده شد")
print(f"📁 مسیر ذخیره: {trainer.save_dir}")

# کامپایل و آموزش مدل
model_cnn.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history = model_cnn.fit(dataset, epochs=5)
```

```python
# ارزیابی مدل جامع
class ModelEvaluator:
    """
    کلاس جامع برای ارزیابی دقیق مدل‌های بینایی ماشین
    """
    
    def __init__(self, model: keras.Model, class_names: List[str] = None):
        """
        مقداردهی اولیه evaluator
        
        Args:
            model: مدل آموزش‌دیده
            class_names: نام کلاس‌ها
        """
        self.model = model
        self.class_names = class_names or [f"Class_{i}" for i in range(10)]
        self.predictions = None
        self.true_labels = None
        
    def evaluate_on_dataset(self, test_dataset, verbose: int = 1) -> Dict:
        """
        ارزیابی جامع روی دیتاست تست
        
        Args:
            test_dataset: دیتاست تست
            verbose: سطح جزئیات
            
        Returns:
            دیکشنری حاوی معیارهای ارزیابی
        """
        print("📊 شروع ارزیابی جامع مدل...")
        
        # ارزیابی اولیه
        test_loss, test_accuracy = self.model.evaluate(test_dataset, verbose=verbose)[:2]
        
        # جمع‌آوری پیش‌بینی‌ها و برچسب‌های واقعی
        self.predictions = []
        self.true_labels = []
        
        print("🔄 جمع‌آوری پیش‌بینی‌ها...")
        for batch_x, batch_y in test_dataset:
            batch_predictions = self.model.predict(batch_x, verbose=0)
            self.predictions.extend(np.argmax(batch_predictions, axis=1))
            self.true_labels.extend(batch_y.numpy())
        
        self.predictions = np.array(self.predictions)
        self.true_labels = np.array(self.true_labels)
        
        # محاسبه معیارهای دقیق
        results = {
            'test_loss': test_loss,
            'test_accuracy': test_accuracy,
            'num_samples': len(self.true_labels),
            'num_classes': len(self.class_names)
        }
        
        # محاسبه معیارهای کلاس‌بندی
        from sklearn.metrics import precision_recall_fscore_support, accuracy_score
        
        precision, recall, f1, support = precision_recall_fscore_support(
            self.true_labels, self.predictions, average=None
        )
        
        # معیارهای میانگین
        macro_precision = np.mean(precision)
        macro_recall = np.mean(recall)
        macro_f1 = np.mean(f1)
        
        weighted_precision = np.average(precision, weights=support)
        weighted_recall = np.average(recall, weights=support)
        weighted_f1 = np.average(f1, weights=support)
        
        results.update({
            'macro_precision': macro_precision,
            'macro_recall': macro_recall,
            'macro_f1': macro_f1,
            'weighted_precision': weighted_precision,
            'weighted_recall': weighted_recall,
            'weighted_f1': weighted_f1,
            'per_class_precision': precision,
            'per_class_recall': recall,
            'per_class_f1': f1,
            'per_class_support': support
        })
        
        print(f"\n✅ ارزیابی تکمیل شد!")
        print(f"📈 دقت تست: {test_accuracy:.4f}")
        print(f"📊 F1-Score (Macro): {macro_f1:.4f}")
        print(f"🎯 F1-Score (Weighted): {weighted_f1:.4f}")
        
        return results
    
    def print_classification_report(self):
        """
        نمایش گزارش تفصیلی کلاس‌بندی
        """
        if self.predictions is None or self.true_labels is None:
            print("❌ ابتدا ارزیابی را اجرا کنید!")
            return
        
        print("\n=== گزارش تفصیلی کلاس‌بندی ===")
        report = classification_report(
            self.true_labels, 
            self.predictions, 
            target_names=self.class_names,
            digits=4
        )
        print(report)
    
    def plot_confusion_matrix(self, normalize: bool = False, figsize: Tuple[int, int] = (10, 8)):
        """
        رسم ماتریس درهم‌ریختگی
        
        Args:
            normalize: نرمال‌سازی ماتریس
            figsize: اندازه نمودار
        """
        if self.predictions is None or self.true_labels is None:
            print("❌ ابتدا ارزیابی را اجرا کنید!")
            return
        
        # محاسبه ماتریس درهم‌ریختگی
        cm = confusion_matrix(self.true_labels, self.predictions)
        
        if normalize:
            cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
            title = 'ماتریس درهم‌ریختگی (نرمال‌شده)'
            fmt = '.2f'
        else:
            title = 'ماتریس درهم‌ریختگی'
            fmt = 'd'
        
        plt.figure(figsize=figsize)
        sns.heatmap(cm, annot=True, fmt=fmt, cmap='Blues',
                    xticklabels=self.class_names, yticklabels=self.class_names)
        plt.title(title)
        plt.xlabel('پیش‌بینی شده')
        plt.ylabel('واقعی')
        plt.tight_layout()
        plt.show()
    
    def plot_per_class_metrics(self, figsize: Tuple[int, int] = (12, 8)):
        """
        رسم معیارهای هر کلاس
        
        Args:
            figsize: اندازه نمودار
        """
        if self.predictions is None or self.true_labels is None:
            print("❌ ابتدا ارزیابی را اجرا کنید!")
            return
        
        from sklearn.metrics import precision_recall_fscore_support
        
        precision, recall, f1, support = precision_recall_fscore_support(
            self.true_labels, self.predictions, average=None
        )
        
        x = np.arange(len(self.class_names))
        width = 0.25
        
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=figsize)
        
        # نمودار معیارها
        ax1.bar(x - width, precision, width, label='Precision', alpha=0.8)
        ax1.bar(x, recall, width, label='Recall', alpha=0.8)
        ax1.bar(x + width, f1, width, label='F1-Score', alpha=0.8)
        
        ax1.set_xlabel('کلاس‌ها')
        ax1.set_ylabel('مقدار معیار')
        ax1.set_title('معیارهای ارزیابی به تفکیک کلاس')
        ax1.set_xticks(x)
        ax1.set_xticklabels(self.class_names, rotation=45)
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # نمودار پشتیبانی (تعداد نمونه‌ها)
        ax2.bar(x, support, alpha=0.7, color='orange')
        ax2.set_xlabel('کلاس‌ها')
        ax2.set_ylabel('تعداد نمونه‌ها')
        ax2.set_title('توزیع نمونه‌ها در کلاس‌ها')
        ax2.set_xticks(x)
        ax2.set_xticklabels(self.class_names, rotation=45)
        ax2.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def analyze_misclassifications(self, top_n: int = 5):
        """
        تحلیل رایج‌ترین اشتباهات کلاس‌بندی
        
        Args:
            top_n: تعداد اشتباهات برتر برای نمایش
        """
        if self.predictions is None or self.true_labels is None:
            print("❌ ابتدا ارزیابی را اجرا کنید!")
            return
        
        print("\n=== تحلیل اشتباهات کلاس‌بندی ===")
        
        # پیدا کردن نمونه‌های اشتباه
        wrong_indices = np.where(self.predictions != self.true_labels)[0]
        
        if len(wrong_indices) == 0:
            print("🎉 هیچ اشتباه کلاس‌بندی وجود ندارد!")
            return
        
        print(f"📊 تعداد کل اشتباهات: {len(wrong_indices)} از {len(self.true_labels)} نمونه")
        print(f"📈 نرخ خطا: {len(wrong_indices)/len(self.true_labels)*100:.2f}%")
        
        # تحلیل الگوهای اشتباه
        error_patterns = {}
        for idx in wrong_indices:
            true_class = self.true_labels[idx]
            pred_class = self.predictions[idx]
            pattern = (true_class, pred_class)
            error_patterns[pattern] = error_patterns.get(pattern, 0) + 1
        
        # مرتب‌سازی براساس فراوانی
        sorted_errors = sorted(error_patterns.items(), key=lambda x: x[1], reverse=True)
        
        print(f"\n🔍 {top_n} الگوی خطای رایج:")
        for i, ((true_cls, pred_cls), count) in enumerate(sorted_errors[:top_n]):
            true_name = self.class_names[true_cls]
            pred_name = self.class_names[pred_cls]
            percentage = count / len(wrong_indices) * 100
            print(f"{i+1}. {true_name} → {pred_name}: {count} بار ({percentage:.1f}% از اشتباهات)")

# ایجاد و استفاده از ModelEvaluator
print("\n=== آماده‌سازی Evaluator جامع ===")

# تعریف نام کلاس‌ها (مثال برای CIFAR-10)
class_names_example = [
    'هواپیما', 'خودرو', 'پرنده', 'گربه', 'آهو',
    'سگ', 'قورباغه', 'اسب', 'کشتی', 'کامیون'
]

evaluator = ModelEvaluator(trainer.model, class_names_example)

print(f"✅ Evaluator برای {len(class_names_example)} کلاس آماده شد")
print("📋 کلاس‌ها:", ', '.join(class_names_example))
```

```python
# رسم نمودار آموزش
import matplotlib.pyplot as plt

class TrainingVisualizer:
    """
    کلاس جامع برای تجسم فرآیند آموزش و نتایج
    """
    
    def __init__(self, history: keras.callbacks.History = None):
        """
        مقداردهی اولیه visualizer
        
        Args:
            history: تاریخچه آموزش
        """
        self.history = history
    
    def plot_training_history(self, figsize: Tuple[int, int] = (15, 5)):
        """
        رسم تاریخچه آموزش (loss و accuracy)
        
        Args:
            figsize: اندازه نمودار
        """
        if self.history is None:
            print("❌ تاریخچه آموزش موجود نیست!")
            return
        
        history_dict = self.history.history
        epochs = range(1, len(history_dict['loss']) + 1)
        
        fig, axes = plt.subplots(1, 3, figsize=figsize)
        
        # نمودار Loss
        axes[0].plot(epochs, history_dict['loss'], 'b-', label='آموزش', linewidth=2)
        if 'val_loss' in history_dict:
            axes[0].plot(epochs, history_dict['val_loss'], 'r-', label='اعتبارسنجی', linewidth=2)
        axes[0].set_title('تغییرات Loss در طول آموزش')
        axes[0].set_xlabel('Epoch')
        axes[0].set_ylabel('Loss')
        axes[0].legend()
        axes[0].grid(True, alpha=0.3)
        
        # نمودار Accuracy
        axes[1].plot(epochs, history_dict['accuracy'], 'b-', label='آموزش', linewidth=2)
        if 'val_accuracy' in history_dict:
            axes[1].plot(epochs, history_dict['val_accuracy'], 'r-', label='اعتبارسنجی', linewidth=2)
        axes[1].set_title('تغییرات Accuracy در طول آموزش')
        axes[1].set_xlabel('Epoch')
        axes[1].set_ylabel('Accuracy')
        axes[1].legend()
        axes[1].grid(True, alpha=0.3)
        
        # نمودار Learning Rate (اگر موجود باشد)
        if 'lr' in history_dict:
            axes[2].plot(epochs, history_dict['lr'], 'g-', linewidth=2)
            axes[2].set_title('تغییرات Learning Rate')
            axes[2].set_xlabel('Epoch')
            axes[2].set_ylabel('Learning Rate')
            axes[2].set_yscale('log')
            axes[2].grid(True, alpha=0.3)
        else:
            # نمودار مقایسه Loss و Accuracy
            ax2_twin = axes[2].twinx()
            line1 = axes[2].plot(epochs, history_dict['loss'], 'b-', label='Loss آموزش', linewidth=2)
            line2 = ax2_twin.plot(epochs, history_dict['accuracy'], 'r-', label='Accuracy آموزش', linewidth=2)
            
            axes[2].set_xlabel('Epoch')
            axes[2].set_ylabel('Loss', color='b')
            ax2_twin.set_ylabel('Accuracy', color='r')
            axes[2].set_title('مقایسه Loss و Accuracy')
            
            # ترکیب legend ها
            lines = line1 + line2
            labels = [l.get_label() for l in lines]
            axes[2].legend(lines, labels, loc='center right')
            axes[2].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        
        # چاپ آمار نهایی
        final_epoch = len(epochs)
        final_train_acc = history_dict['accuracy'][-1]
        final_train_loss = history_dict['loss'][-1]
        
        print(f"\n📊 آمار نهایی آموزش:")
        print(f"🔢 تعداد epoch های اجرا شده: {final_epoch}")
        print(f"📈 دقت نهایی آموزش: {final_train_acc:.4f}")
        print(f"📉 خطای نهایی آموزش: {final_train_loss:.4f}")
        
        if 'val_accuracy' in history_dict and 'val_loss' in history_dict:
            final_val_acc = history_dict['val_accuracy'][-1]
            final_val_loss = history_dict['val_loss'][-1]
            print(f"📈 دقت نهایی اعتبارسنجی: {final_val_acc:.4f}")
            print(f"📉 خطای نهایی اعتبارسنجی: {final_val_loss:.4f}")
            
            # تشخیص overfitting
            train_val_diff = final_train_acc - final_val_acc
            if train_val_diff > 0.05:  # اختلاف بیش از 5%
                print(f"⚠️ احتمال overfitting: اختلاف دقت آموزش و اعتبارسنجی = {train_val_diff:.3f}")
            else:
                print(f"✅ مدل متعادل: اختلاف دقت آموزش و اعتبارسنجی = {train_val_diff:.3f}")
    
    def plot_learning_curves(self, train_sizes: np.ndarray = None, 
                           train_scores: np.ndarray = None,
                           val_scores: np.ndarray = None,
                           figsize: Tuple[int, int] = (10, 6)):
        """
        رسم منحنی‌های یادگیری برای تشخیص overfitting/underfitting
        
        Args:
            train_sizes: اندازه‌های مختلف دیتاست آموزش
            train_scores: امتیازات آموزش
            val_scores: امتیازات اعتبارسنجی
            figsize: اندازه نمودار
        """
        if train_sizes is None or train_scores is None or val_scores is None:
            print("❌ داده‌های منحنی یادگیری موجود نیست!")
            return
        
        plt.figure(figsize=figsize)
        
        # محاسبه میانگین و انحراف معیار
        train_mean = np.mean(train_scores, axis=1)
        train_std = np.std(train_scores, axis=1)
        val_mean = np.mean(val_scores, axis=1)
        val_std = np.std(val_scores, axis=1)
        
        # رسم منحنی‌ها
        plt.plot(train_sizes, train_mean, 'o-', color='blue', label='امتیاز آموزش')
        plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.1, color='blue')
        
        plt.plot(train_sizes, val_mean, 'o-', color='red', label='امتیاز اعتبارسنجی')
        plt.fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.1, color='red')
        
        plt.xlabel('تعداد نمونه‌های آموزش')
        plt.ylabel('امتیاز')
        plt.title('منحنی‌های یادگیری')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
    
    def plot_metric_comparison(self, metrics_dict: Dict[str, List[float]], 
                             figsize: Tuple[int, int] = (12, 8)):
        """
        مقایسه معیارهای مختلف در یک نمودار
        
        Args:
            metrics_dict: دیکشنری شامل نام معیار و مقادیر آن
            figsize: اندازه نمودار
        """
        plt.figure(figsize=figsize)
        
        epochs = range(1, len(list(metrics_dict.values())[0]) + 1)
        
        for metric_name, values in metrics_dict.items():
            plt.plot(epochs, values, 'o-', label=metric_name, linewidth=2, markersize=4)
        
        plt.xlabel('Epoch')
        plt.ylabel('مقدار معیار')
        plt.title('مقایسه معیارهای مختلف')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
    
    def create_training_summary(self) -> str:
        """
        ایجاد خلاصه متنی از فرآیند آموزش
        
        Returns:
            خلاصه متنی آموزش
        """
        if self.history is None:
            return "❌ تاریخچه آموزش موجود نیست!"
        
        history_dict = self.history.history
        total_epochs = len(history_dict['loss'])
        
        # بهترین epoch ها
        best_train_acc_epoch = np.argmax(history_dict['accuracy']) + 1
        best_train_acc = np.max(history_dict['accuracy'])
        
        summary = f"""
📋 === خلاصه آموزش مدل ===

🔢 تعداد کل epoch ها: {total_epochs}
📈 بهترین دقت آموزش: {best_train_acc:.4f} (epoch {best_train_acc_epoch})
📉 خطای نهایی آموزش: {history_dict['loss'][-1]:.4f}
"""
        
        if 'val_accuracy' in history_dict:
            best_val_acc_epoch = np.argmax(history_dict['val_accuracy']) + 1
            best_val_acc = np.max(history_dict['val_accuracy'])
            summary += f"""
📈 بهترین دقت اعتبارسنجی: {best_val_acc:.4f} (epoch {best_val_acc_epoch})
📉 خطای نهایی اعتبارسنجی: {history_dict['val_loss'][-1]:.4f}
"""
            
            # تحلیل پایداری
            if best_val_acc_epoch < total_epochs * 0.8:
                summary += "\n⚠️ توصیه: احتمال overfitting - در نظر گیرید: Early Stopping، Regularization"
            else:
                summary += "\n✅ مدل پایدار: عملکرد مناسب بدون overfitting واضح"
        
        return summary

# ایجاد visualizer و نمایش نمونه
print("\n=== آماده‌سازی Visualizer ===")

# ایجاد تاریخچه نمونه برای تست
sample_history = keras.callbacks.History()
sample_history.history = {
    'loss': [2.3, 1.8, 1.4, 1.1, 0.9, 0.7, 0.6, 0.5],
    'accuracy': [0.1, 0.3, 0.5, 0.7, 0.8, 0.85, 0.88, 0.90],
    'val_loss': [2.2, 1.9, 1.5, 1.3, 1.1, 1.0, 0.95, 0.92],
    'val_accuracy': [0.15, 0.35, 0.52, 0.68, 0.75, 0.79, 0.82, 0.84]
}

visualizer = TrainingVisualizer(sample_history)

print("✅ Visualizer آماده شد")
print("📊 نمونه تجسم تاریخچه آموزش:")
visualizer.plot_training_history()

print("\n📋 خلاصه آموزش:")
print(visualizer.create_training_summary())
```

## ۵.۲. استراتژی‌های پیشرفته آموزش

### Hyperparameter Tuning:

- **Random Search**: جستجوی تصادفی
- **Bayesian Optimization**: بهینه‌سازی بیزی
- **Hyperband**: روش ترکیبی سریع

### تکنیک‌های Regularization:
- **L1/L2 Regularization**: جریمه وزن‌ها
- **Dropout**: حذف تصادفی نورون‌ها
- **Batch Normalization**: نرمال‌سازی batch ها
- **Early Stopping**: توقف زودهنگام
- **Data Augmentation**: افزایش مصنوعی داده

### Learning Rate Scheduling:
- **Step Decay**: کاهش پله‌ای
- **Exponential Decay**: کاهش نمایی
- **Cosine Annealing**: تغییر کسینوسی
- **Cyclical Learning Rates**: نرخ چرخشی

```python
class HyperparameterTuner:
    """
    کلاس تنظیم خودکار هایپرپارامترها
    """
    
    def __init__(self, model_builder_func, param_space: Dict):
        """
        مقداردهی اولیه tuner
        
        Args:
            model_builder_func: تابع سازنده مدل
            param_space: فضای جستجوی پارامترها
        """
        self.model_builder = model_builder_func
        self.param_space = param_space
        self.results = []
    
    def random_search(self, train_dataset, val_dataset, 
                     n_trials: int = 10, epochs: int = 5) -> Dict:
        """
        جستجوی تصادفی هایپرپارامترها
        
        Args:
            train_dataset: دیتاست آموزش
            val_dataset: دیتاست اعتبارسنجی
            n_trials: تعداد آزمایش
            epochs: تعداد epoch برای هر آزمایش
            
        Returns:
            بهترین پارامترها
        """
        print(f"🔍 شروع جستجوی تصادفی با {n_trials} آزمایش")
        
        best_score = 0
        best_params = None
        
        for trial in range(n_trials):
            print(f"\n🧪 آزمایش {trial + 1}/{n_trials}")
            
            # انتخاب تصادفی پارامترها
            params = {}
            for param_name, param_values in self.param_space.items():
                if isinstance(param_values, list):
                    params[param_name] = np.random.choice(param_values)
                elif isinstance(param_values, tuple) and len(param_values) == 2:
                    if isinstance(param_values[0], float):
                        params[param_name] = np.random.uniform(param_values[0], param_values[1])
                    else:
                        params[param_name] = np.random.randint(param_values[0], param_values[1] + 1)
            
            print(f"📊 پارامترهای انتخاب شده: {params}")
            
            try:
                # ساخت و آموزش مدل
                model = self.model_builder(**params)
                
                # کامپایل مدل
                model.compile(
                    optimizer=keras.optimizers.Adam(learning_rate=params.get('learning_rate', 0.001)),
                    loss='sparse_categorical_crossentropy',
                    metrics=['accuracy']
                )
                
                # آموزش مدل
                history = model.fit(
                    train_dataset,
                    validation_data=val_dataset,
                    epochs=epochs,
                    verbose=0
                )
                
                # ارزیابی
                val_accuracy = max(history.history['val_accuracy'])
                
                result = {
                    'trial': trial + 1,
                    'params': params.copy(),
                    'val_accuracy': val_accuracy,
                    'history': history.history
                }
                
                self.results.append(result)
                
                print(f"📈 دقت اعتبارسنجی: {val_accuracy:.4f}")
                
                # به‌روزرسانی بهترین نتیجه
                if val_accuracy > best_score:
                    best_score = val_accuracy
                    best_params = params.copy()
                    print(f"🎯 بهترین نتیجه جدید!")
                
            except Exception as e:
                print(f"❌ خطا در آزمایش: {e}")
                continue
        
        print(f"\n✅ جستجو تکمیل شد!")
        print(f"🏆 بهترین دقت: {best_score:.4f}")
        print(f"🔧 بهترین پارامترها: {best_params}")
        
        return {
            'best_params': best_params,
            'best_score': best_score,
            'all_results': self.results
        }
    
    def plot_tuning_results(self, figsize: Tuple[int, int] = (12, 8)):
        """
        تجسم نتایج تنظیم هایپرپارامترها
        
        Args:
            figsize: اندازه نمودار
        """
        if not self.results:
            print("❌ نتیجه‌ای برای نمایش وجود ندارد!")
            return
        
        trials = [r['trial'] for r in self.results]
        scores = [r['val_accuracy'] for r in self.results]
        
        plt.figure(figsize=figsize)
        
        # نمودار امتیازات
        plt.subplot(1, 2, 1)
        plt.plot(trials, scores, 'bo-', linewidth=2, markersize=8)
        plt.axhline(y=max(scores), color='r', linestyle='--', alpha=0.7, label=f'بهترین: {max(scores):.4f}')
        plt.xlabel('شماره آزمایش')
        plt.ylabel('دقت اعتبارسنجی')
        plt.title('نتایج تنظیم هایپرپارامترها')
        plt.legend()
        plt.grid(True, alpha=0.3)
        
        # هیستوگرام امتیازات
        plt.subplot(1, 2, 2)
        plt.hist(scores, bins=min(10, len(scores)), alpha=0.7, color='skyblue', edgecolor='black')
        plt.axvline(x=max(scores), color='r', linestyle='--', alpha=0.7, label=f'بهترین: {max(scores):.4f}')
        plt.xlabel('دقت اعتبارسنجی')
        plt.ylabel('فراوانی')
        plt.title('توزیع امتیازات')
        plt.legend()
        plt.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()

# مثال استفاده از HyperparameterTuner
print("\n=== مثال تنظیم هایپرپارامترها ===")

# تعریف تابع سازنده مدل
def build_cnn_model(filters1=32, filters2=64, dense_units=64, dropout_rate=0.5):
    """
    تابع سازنده مدل CNN با پارامترهای قابل تنظیم
    """
    model = keras.Sequential([
        layers.Conv2D(filters1, 3, activation='relu', input_shape=(32, 32, 3)),
        layers.MaxPooling2D(),
        layers.Conv2D(filters2, 3, activation='relu'),
        layers.MaxPooling2D(),
        layers.Flatten(),
        layers.Dense(dense_units, activation='relu'),
        layers.Dropout(dropout_rate),
        layers.Dense(10, activation='softmax')
    ])
    return model

# تعریف فضای جستجو
param_space = {
    'filters1': [16, 32, 64],
    'filters2': [32, 64, 128],
    'dense_units': [32, 64, 128],
    'dropout_rate': [0.2, 0.3, 0.5],
    'learning_rate': (0.0001, 0.01)  # بازه پیوسته
}

tuner = HyperparameterTuner(build_cnn_model, param_space)

print("✅ Tuner آماده شد")
print(f"🔧 فضای جستجو: {param_space}")
print("💡 برای اجرای جستجو، از روش random_search استفاده کنید")
```

### پروژه ۳: سگمنتیشن تصاویر با U-Net

در این پروژه یک مدل U-Net برای سگمنتیشن تصاویر پزشکی پیاده‌سازی می‌شود.

```python
# مدل U-Net ساده
def unet_model(input_shape=(128,128,3), num_classes=2):
    inputs = tf.keras.Input(shape=input_shape)
    c1 = tf.keras.layers.Conv2D(16, 3, activation='relu', padding='same')(inputs)
    c1 = tf.keras.layers.Conv2D(16, 3, activation='relu', padding='same')(c1)
    p1 = tf.keras.layers.MaxPooling2D()(c1)
    c2 = tf.keras.layers.Conv2D(32, 3, activation='relu', padding='same')(p1)
    c2 = tf.keras.layers.Conv2D(32, 3, activation='relu', padding='same')(c2)
    p2 = tf.keras.layers.MaxPooling2D()(c2)
    c3 = tf.keras.layers.Conv2D(64, 3, activation='relu', padding='same')(p2)
    c3 = tf.keras.layers.Conv2D(64, 3, activation='relu', padding='same')(c3)
    u1 = tf.keras.layers.UpSampling2D()(c3)
    concat1 = tf.keras.layers.Concatenate()([u1, c2])
    c4 = tf.keras.layers.Conv2D(32, 3, activation='relu', padding='same')(concat1)
    u2 = tf.keras.layers.UpSampling2D()(c4)
    concat2 = tf.keras.layers.Concatenate()([u2, c1])
    c5 = tf.keras.layers.Conv2D(16, 3, activation='relu', padding='same')(concat2)
    outputs = tf.keras.layers.Conv2D(num_classes, 1, activation='softmax')(c5)
    return tf.keras.Model(inputs, outputs)

model_unet = unet_model()
model_unet.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model_unet.summary()
```

### ۲.۴ متغیرها و محاسبه گرادیان

در یادگیری عمیق، نیاز به متغیرهایی داریم که قابل آموزش باشند و بتوانیم گرادیان آنها را محاسبه کنیم.

```python
# ۲.۴.۱ متغیرهای TensorFlow

# ایجاد متغیر
weight = tf.Variable(tf.random.normal((3, 2)), name='weight')
bias = tf.Variable(tf.zeros((2,)), name='bias')

print("وزن اولیه:")
print(weight)
print("نوع:", type(weight))
print("قابل آموزش:", weight.trainable)
print()
print("بایاس اولیه:")
print(bias)
print()

# محاسبه گرادیان با GradientTape
x = tf.constant([[1.0, 2.0, 3.0]])

with tf.GradientTape() as tape:
    # محاسبه ساده: y = x @ weight + bias
    y = tf.matmul(x, weight) + bias
    # تابع هزینه ساده
    loss = tf.reduce_mean(tf.square(y))

# محاسبه گرادیان
grads = tape.gradient(loss, [weight, bias])
weight_grad, bias_grad = grads

print(f"هزینه: {loss.numpy():.4f}")
print("گرادیان وزن:")
print(weight_grad)
print("گرادیان بایاس:")
print(bias_grad)
print()

# به‌روزرسانی متغیرها (شبیه‌سازی یک قدم آموزش)
learning_rate = 0.1
weight.assign_sub(learning_rate * weight_grad)
bias.assign_sub(learning_rate * bias_grad)

print("وزن بعد از به‌روزرسانی:")
print(weight)
print("بایاس بعد از به‌روزرسانی:")
print(bias)
```

### ۲.۶ مثال عملی: پیش‌پردازش تصویر

حالا یک مثال کامل از پیش‌پردازش تصویر برای بینایی ماشین خواهیم دید.

```python
# ۲.۶.۱ مثال کامل پیش‌پردازش تصویر

def preprocess_image(image, target_size=(224, 224), normalize=True):
    """
    تابع پیش‌پردازش کامل تصویر برای مدل‌های یادگیری عمیق
    
    Args:
        image: تصویر ورودی (Tensor یا NumPy array)
        target_size: اندازه هدف (tuple)
        normalize: آیا نرمال‌سازی انجام شود یا خیر
    
    Returns:
        تصویر پردازش‌شده
    """
    # اطمینان از نوع داده مناسب
    if isinstance(image, np.ndarray):
        image = tf.constant(image)
    
    # تبدیل به float32 اگر نیاز باشد
    if image.dtype != tf.float32:
        image = tf.cast(image, tf.float32)
    
    # نرمال‌سازی اگر در محدوده 0-255 باشد
    if tf.reduce_max(image) > 1.0:
        image = image / 255.0
    
    # تغییر اندازه
    image = tf.image.resize(image, target_size)
    
    # نرمال‌سازی با میانگین و انحراف معیار ImageNet (اختیاری)
    if normalize:
        mean = tf.constant([0.485, 0.456, 0.406])
        std = tf.constant([0.229, 0.224, 0.225])
        image = (image - mean) / std
    
    return image

# تست تابع پیش‌پردازش
test_image = tf.random.uniform((150, 200, 3), minval=0, maxval=255)
print("تصویر اصلی:")
print(f"شکل: {test_image.shape}")
print(f"محدوده: {tf.reduce_min(test_image):.2f} تا {tf.reduce_max(test_image):.2f}")
print()

# پیش‌پردازش
processed = preprocess_image(test_image)
print("تصویر پردازش‌شده:")
print(f"شکل: {processed.shape}")
print(f"محدوده: {tf.reduce_min(processed):.3f} تا {tf.reduce_max(processed):.3f}")
print(f"میانگین: {tf.reduce_mean(processed):.3f}")
print(f"انحراف معیار: {tf.math.reduce_std(processed):.3f}")
print()

# پردازش دسته‌ای (Batch Processing)
batch_images = tf.random.uniform((5, 180, 240, 3), minval=0, maxval=255)
print("دسته تصاویر اصلی:", batch_images.shape)

# پردازش هر تصویر در دسته
processed_batch = tf.stack([
    preprocess_image(img) for img in tf.unstack(batch_images)
])
print("دسته پردازش‌شده:", processed_batch.shape)

# یا استفاده از map_fn برای کارایی بهتر
processed_batch_efficient = tf.map_fn(
    lambda img: preprocess_image(img),
    batch_images,
    dtype=tf.float32
)
print("دسته پردازش‌شده (کارآمد):", processed_batch_efficient.shape)
```

## ۳.۱ بارگذاری و خواندن تصاویر

TensorFlow روش‌های مختلفی برای بارگذاری تصاویر ارائه می‌دهد که هر کدام برای موارد استفاده خاصی بهینه شده‌اند.

### روش‌های مختلف بارگذاری تصاویر:

1. **tf.io**: برای خواندن فایل‌های تصویری از دیسک
2. **tf.keras.utils.load_img**: برای بارگذاری ساده تصاویر
3. **tf.data.Dataset**: برای پردازش batch و pipeline
4. **tf.image**: برای عملیات پردازش تصویر
5. **OpenCV + TensorFlow**: ترکیب OpenCV با TensorFlow

### مزایای هر روش:
- **tf.io**: سریع و مناسب برای فایل‌های بزرگ
- **keras.utils**: ساده و مناسب برای شروع
- **tf.data**: بهینه برای training pipeline
- **tf.image**: عملیات پیشرفته روی تصاویر
- **OpenCV**: قدرت بالا در پردازش تصویر

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img, img_to_array
import os
import cv2
from pathlib import Path

# تنظیم matplotlib برای نمایش فونت فارسی
plt.rcParams['font.family'] = ['Arial Unicode MS', 'Tahoma', 'DejaVu Sans']

print("TensorFlow version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))

# ۱. بارگذاری تصویر با tf.io
def load_image_with_tf_io(image_path, target_size=(224, 224)):
    """
    بارگذاری تصویر با استفاده از tf.io
    
    Args:
        image_path: مسیر فایل تصویر
        target_size: اندازه هدف (height, width)
    
    Returns:
        tensor تصویر با شکل (height, width, channels)
    """
    # خواندن فایل
    image_raw = tf.io.read_file(image_path)
    
    # تشخیص نوع فایل و decode کردن
    if tf.strings.lower(tf.strings.substr(image_path, -4, 4)) == '.png':
        image = tf.image.decode_png(image_raw, channels=3)
    else:
        image = tf.image.decode_jpeg(image_raw, channels=3)
    
    # تبدیل به float32 و نرمال‌سازی
    image = tf.cast(image, tf.float32) / 255.0
    
    # تغییر اندازه
    image = tf.image.resize(image, target_size)
    
    return image

# ۲. بارگذاری تصویر با Keras
def load_image_with_keras(image_path, target_size=(224, 224)):
    """
    بارگذاری تصویر با استفاده از Keras utilities
    """
    # بارگذاری تصویر
    image = load_img(image_path, target_size=target_size)
    
    # تبدیل به array
    image_array = img_to_array(image)
    
    # نرمال‌سازی
    image_array = image_array / 255.0
    
    return image_array

# ۳. بارگذاری با OpenCV
def load_image_with_opencv(image_path, target_size=(224, 224)):
    """
    بارگذاری تصویر با استفاده از OpenCV
    """
    # خواندن تصویر (BGR format)
    image = cv2.imread(image_path)
    
    if image is None:
        raise ValueError(f"نمی‌توان تصویر را از مسیر {image_path} بارگذاری کرد")
    
    # تبدیل BGR به RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # تغییر اندازه
    image = cv2.resize(image, target_size)
    
    # نرمال‌سازی
    image = image.astype(np.float32) / 255.0
    
    return image

# ۴. تابع جامع بارگذاری تصویر
def load_and_preprocess_image(image_path, target_size=(224, 224), method='tf_io'):
    """
    تابع جامع برای بارگذاری و پیش‌پردازش تصویر
    
    Args:
        image_path: مسیر تصویر
        target_size: اندازه هدف
        method: روش بارگذاری ('tf_io', 'keras', 'opencv')
    
    Returns:
        تصویر پردازش شده
    """
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"فایل تصویر در مسیر {image_path} یافت نشد")
    
    if method == 'tf_io':
        return load_image_with_tf_io(image_path, target_size)
    elif method == 'keras':
        return load_image_with_keras(image_path, target_size)
    elif method == 'opencv':
        return load_image_with_opencv(image_path, target_size)
    else:
        raise ValueError("روش باید یکی از 'tf_io', 'keras', 'opencv' باشد")

# تست مثال (در صورت وجود تصویر)
image_path = "../images/Lenna.png"  # مسیر تصویر نمونه

if os.path.exists(image_path):
    print("\n=== تست بارگذاری تصویر ===")
    
    # تست روش‌های مختلف
    methods = ['tf_io', 'keras', 'opencv']
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    for i, method in enumerate(methods):
        try:
            image = load_and_preprocess_image(image_path, method=method)
            axes[i].imshow(image)
            axes[i].set_title(f'روش: {method}\nشکل: {image.shape}')
            axes[i].axis('off')
            print(f"✅ {method}: شکل = {image.shape}, نوع = {type(image)}")
        except Exception as e:
            print(f"❌ خطا در {method}: {e}")
            axes[i].text(0.5, 0.5, f'خطا در {method}', 
                        ha='center', va='center', transform=axes[i].transAxes)
    
    plt.tight_layout()
    plt.show()
else:
    print(f"⚠️ فایل تصویر در مسیر {image_path} یافت نشد")
    print("برای تست، یک تصویر در مسیر images قرار دهید")

print("\n✅ بخش بارگذاری تصاویر تکمیل شد")
```

## ۳.۲ پیش‌پردازش تصاویر

پیش‌پردازش تصاویر یکی از مهم‌ترین مراحل در computer vision است که کیفیت و دقت مدل را به شدت تحت تأثیر قرار می‌دهد.

### عملیات‌های اصلی پیش‌پردازش:

1. **تغییر اندازه (Resizing)**: تطبیق اندازه تصاویر با ورودی مدل
2. **نرمال‌سازی (Normalization)**: تنظیم مقادیر پیکسل‌ها
3. **تبدیل فضای رنگ (Color Space Conversion)**: RGB, BGR, Grayscale, HSV
4. **Padding و Cropping**: تنظیم نسبت ابعاد
5. **تصحیح نور (Brightness/Contrast)**: بهبود کیفیت تصویر
6. **حذف نویز (Denoising)**: بهبود وضوح تصویر

### استراتژی‌های نرمال‌سازی:

- **Min-Max Scaling**: (0, 1) یا (-1, 1)
- **Z-Score Normalization**: میانگین=0، انحراف معیار=1
- **ImageNet Normalization**: استفاده از آمار ImageNet
- **Per-Channel Normalization**: نرمال‌سازی جداگانه هر کانال

```python
# کلاس جامع برای پیش‌پردازش تصاویر
class ImagePreprocessor:
    """
    کلاس جامع برای پیش‌پردازش تصاویر در TensorFlow
    """
    
    def __init__(self, target_size=(224, 224), normalization='minmax'):
        self.target_size = target_size
        self.normalization = normalization
        
        # آمار ImageNet برای نرمال‌سازی
        self.imagenet_mean = [0.485, 0.456, 0.406]
        self.imagenet_std = [0.229, 0.224, 0.225]
    
    def resize_image(self, image, method='bilinear', preserve_aspect_ratio=False):
        """
        تغییر اندازه تصویر با روش‌های مختلف
        
        Args:
            image: تصویر ورودی
            method: روش تغییر اندازه ('bilinear', 'nearest', 'bicubic')
            preserve_aspect_ratio: حفظ نسبت ابعاد
        """
        if preserve_aspect_ratio:
            # محاسبه اندازه جدید با حفظ نسبت
            height, width = tf.shape(image)[0], tf.shape(image)[1]
            target_h, target_w = self.target_size
            
            # محاسبه scale factor
            scale = tf.minimum(
                tf.cast(target_h, tf.float32) / tf.cast(height, tf.float32),
                tf.cast(target_w, tf.float32) / tf.cast(width, tf.float32)
            )
            
            new_height = tf.cast(tf.cast(height, tf.float32) * scale, tf.int32)
            new_width = tf.cast(tf.cast(width, tf.float32) * scale, tf.int32)
            
            # تغییر اندازه
            if method == 'bilinear':
                resized = tf.image.resize(image, [new_height, new_width], 
                                        method=tf.image.ResizeMethod.BILINEAR)
            elif method == 'nearest':
                resized = tf.image.resize(image, [new_height, new_width], 
                                        method=tf.image.ResizeMethod.NEAREST_NEIGHBOR)
            elif method == 'bicubic':
                resized = tf.image.resize(image, [new_height, new_width], 
                                        method=tf.image.ResizeMethod.BICUBIC)
            
            # Padding برای رسیدن به اندازه هدف
            padded = tf.image.resize_with_pad(resized, target_h, target_w)
            return padded
        else:
            # تغییر اندازه مستقیم
            if method == 'bilinear':
                return tf.image.resize(image, self.target_size, 
                                     method=tf.image.ResizeMethod.BILINEAR)
            elif method == 'nearest':
                return tf.image.resize(image, self.target_size, 
                                     method=tf.image.ResizeMethod.NEAREST_NEIGHBOR)
            elif method == 'bicubic':
                return tf.image.resize(image, self.target_size, 
                                     method=tf.image.ResizeMethod.BICUBIC)
    
    def normalize_image(self, image, method=None):
        """
        نرمال‌سازی تصویر با روش‌های مختلف
        """
        if method is None:
            method = self.normalization
        
        # تبدیل به float32
        image = tf.cast(image, tf.float32)
        
        if method == 'minmax':
            # نرمال‌سازی به بازه [0, 1]
            return image / 255.0
        
        elif method == 'minmax_centered':
            # نرمال‌سازی به بازه [-1, 1]
            return (image / 127.5) - 1.0
        
        elif method == 'zscore':
            # Z-Score normalization
            mean = tf.reduce_mean(image)
            std = tf.math.reduce_std(image)
            return (image - mean) / (std + 1e-7)
        
        elif method == 'imagenet':
            # نرمال‌سازی ImageNet
            image = image / 255.0  # ابتدا به [0,1]
            mean = tf.constant(self.imagenet_mean, dtype=tf.float32)
            std = tf.constant(self.imagenet_std, dtype=tf.float32)
            return (image - mean) / std
        
        elif method == 'per_channel':
            # نرمال‌سازی جداگانه هر کانال
            image = image / 255.0
            mean = tf.reduce_mean(image, axis=[0, 1], keepdims=True)
            std = tf.math.reduce_std(image, axis=[0, 1], keepdims=True)
            return (image - mean) / (std + 1e-7)
        
        else:
            raise ValueError(f"روش نرمال‌سازی '{method}' پشتیبانی نمی‌شود")
    
    def convert_color_space(self, image, target_space='rgb'):
        """
        تبدیل فضای رنگ تصویر
        """
        if target_space == 'rgb':
            return image  # فرض: تصویر از قبل RGB است
        elif target_space == 'grayscale':
            return tf.image.rgb_to_grayscale(image)
        elif target_space == 'hsv':
            return tf.image.rgb_to_hsv(image)
        elif target_space == 'yuv':
            return tf.image.rgb_to_yuv(image)
        else:
            raise ValueError(f"فضای رنگ '{target_space}' پشتیبانی نمی‌شود")
    
    def adjust_brightness_contrast(self, image, brightness_delta=0.0, contrast_factor=1.0):
        """
        تنظیم نور و کنتراست تصویر
        """
        # تنظیم نور
        if brightness_delta != 0.0:
            image = tf.image.adjust_brightness(image, brightness_delta)
        
        # تنظیم کنتراست
        if contrast_factor != 1.0:
            image = tf.image.adjust_contrast(image, contrast_factor)
        
        # کلمپ کردن مقادیر
        return tf.clip_by_value(image, 0.0, 1.0)
    
    def add_noise(self, image, noise_type='gaussian', noise_factor=0.1):
        """
        اضافه کردن نویز به تصویر
        """
        if noise_type == 'gaussian':
            noise = tf.random.normal(tf.shape(image), mean=0.0, stddev=noise_factor)
            return tf.clip_by_value(image + noise, 0.0, 1.0)
        elif noise_type == 'uniform':
            noise = tf.random.uniform(tf.shape(image), -noise_factor, noise_factor)
            return tf.clip_by_value(image + noise, 0.0, 1.0)
        else:
            raise ValueError(f"نوع نویز '{noise_type}' پشتیبانی نمی‌شود")
    
    def preprocess(self, image, **kwargs):
        """
        پیش‌پردازش کامل تصویر
        """
        # تغییر اندازه
        image = self.resize_image(image, 
                                method=kwargs.get('resize_method', 'bilinear'),
                                preserve_aspect_ratio=kwargs.get('preserve_aspect_ratio', False))
        
        # تبدیل فضای رنگ
        if 'color_space' in kwargs:
            image = self.convert_color_space(image, kwargs['color_space'])
        
        # تنظیم نور و کنتراست
        if 'brightness_delta' in kwargs or 'contrast_factor' in kwargs:
            image = self.adjust_brightness_contrast(
                image, 
                kwargs.get('brightness_delta', 0.0),
                kwargs.get('contrast_factor', 1.0)
            )
        
        # اضافه کردن نویز
        if 'noise_type' in kwargs:
            image = self.add_noise(image, 
                                 kwargs['noise_type'], 
                                 kwargs.get('noise_factor', 0.1))
        
        # نرمال‌سازی
        image = self.normalize_image(image, kwargs.get('normalization_method'))
        
        return image

# تست پیش‌پردازش
print("=== تست پیش‌پردازش تصاویر ===")

# ایجاد تصویر نمونه
sample_image = tf.random.uniform((300, 400, 3), 0, 255, dtype=tf.float32)
print(f"تصویر نمونه: شکل = {sample_image.shape}")

# ایجاد preprocessor
preprocessor = ImagePreprocessor(target_size=(224, 224), normalization='minmax')

# تست روش‌های مختلف نرمال‌سازی
normalization_methods = ['minmax', 'minmax_centered', 'zscore', 'imagenet', 'per_channel']

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

# تصویر اصلی
axes[0].imshow(sample_image.numpy().astype(np.uint8))
axes[0].set_title('تصویر اصلی')
axes[0].axis('off')

# تست روش‌های نرمال‌سازی
for i, method in enumerate(normalization_methods):
    try:
        processed = preprocessor.preprocess(sample_image, normalization_method=method)
        
        # برای نمایش، مقادیر را به [0,1] تبدیل می‌کنیم
        display_image = processed
        if method == 'minmax_centered':
            display_image = (processed + 1.0) / 2.0
        elif method in ['zscore', 'imagenet', 'per_channel']:
            display_image = tf.clip_by_value((processed - tf.reduce_min(processed)) / 
                                           (tf.reduce_max(processed) - tf.reduce_min(processed)), 0, 1)
        
        axes[i+1].imshow(display_image.numpy())
        axes[i+1].set_title(f'{method}\nMin: {tf.reduce_min(processed):.3f}, Max: {tf.reduce_max(processed):.3f}')
        axes[i+1].axis('off')
        print(f"✅ {method}: شکل = {processed.shape}, بازه = [{tf.reduce_min(processed):.3f}, {tf.reduce_max(processed):.3f}]")
    except Exception as e:
        print(f"❌ خطا در {method}: {e}")
        axes[i+1].text(0.5, 0.5, f'خطا در {method}', ha='center', va='center', transform=axes[i+1].transAxes)

plt.tight_layout()
plt.show()

print("\n✅ بخش پیش‌پردازش تصاویر تکمیل شد")
```

## ۳.۳ Data Augmentation (افزایش داده)

Data Augmentation یکی از تکنیک‌های اساسی برای بهبود عملکرد مدل‌های یادگیری عمیق است که با ایجاد تنوع در داده‌های آموزشی، مدل را مقاوم‌تر می‌کند.

### انواع تبدیلات Data Augmentation:

#### ۱. تبدیلات هندسی (Geometric Transforms):
- چرخش (Rotation)
- انتقال (Translation) 
- مقیاس‌گذاری (Scaling)
- برش (Cropping)
- انعکاس (Flipping)
- برش تصادفی (Random Crop)

#### ۲. تبدیلات رنگی (Color Transforms):
- تغییر روشنایی (Brightness)
- تغییر کنتراست (Contrast)
- تغییر اشباع رنگ (Saturation)
- تغییر رنگ (Hue)
- اعمال فیلتر (Color Jittering)

#### ۳. تبدیلات پیشرفته:
- Cutout/Random Erasing
- MixUp
- CutMix
- Gaussian Blur
- Elastic Deformation

### مزایای Data Augmentation:
- کاهش Overfitting
- افزایش مقاومت مدل
- بهبود تعمیم‌پذیری
- افزایش حجم داده‌های آموزشی

```python
# کلاس Data Augmentation برای TensorFlow
class DataAugmentator:
    """
    کلاس جامع برای اعمال Data Augmentation بر روی تصاویر
    """
    
    def __init__(self, seed=42):
        self.seed = seed
        tf.random.set_seed(seed)
    
    def random_flip(self, image, mode='horizontal'):
        """
        انعکاس تصادفی تصویر
        """
        if mode == 'horizontal':
            return tf.image.random_flip_left_right(image)
        elif mode == 'vertical':
            return tf.image.random_flip_up_down(image)
        elif mode == 'both':
            image = tf.image.random_flip_left_right(image)
            image = tf.image.random_flip_up_down(image)
            return image
        else:
            raise ValueError("mode باید 'horizontal', 'vertical' یا 'both' باشد")
    
    def random_rotation(self, image, max_angle=30):
        """
        چرخش تصادفی تصویر
        """
        # تبدیل درجه به رادیان
        angle = tf.random.uniform([], -max_angle, max_angle) * tf.constant(np.pi) / 180.0
        return tf.contrib.image.rotate(image, angle)
    
    def random_crop(self, image, crop_size):
        """
        برش تصادفی از تصویر
        """
        if len(crop_size) == 2:
            crop_size = (*crop_size, tf.shape(image)[-1])
        return tf.image.random_crop(image, crop_size)
    
    def random_brightness(self, image, max_delta=0.2):
        """
        تغییر تصادفی روشنایی
        """
        return tf.image.random_brightness(image, max_delta)
    
    def random_contrast(self, image, lower=0.8, upper=1.2):
        """
        تغییر تصادفی کنتراست
        """
        return tf.image.random_contrast(image, lower, upper)
    
    def random_saturation(self, image, lower=0.8, upper=1.2):
        """
        تغییر تصادفی اشباع رنگ
        """
        return tf.image.random_saturation(image, lower, upper)
    
    def random_hue(self, image, max_delta=0.1):
        """
        تغییر تصادفی رنگ
        """
        return tf.image.random_hue(image, max_delta)
    
    def color_jitter(self, image, brightness=0.2, contrast=(0.8, 1.2), 
                    saturation=(0.8, 1.2), hue=0.1):
        """
        اعمال همزمان تغییرات رنگی
        """
        if brightness > 0:
            image = self.random_brightness(image, brightness)
        if contrast[0] != 1.0 or contrast[1] != 1.0:
            image = self.random_contrast(image, contrast[0], contrast[1])
        if saturation[0] != 1.0 or saturation[1] != 1.0:
            image = self.random_saturation(image, saturation[0], saturation[1])
        if hue > 0:
            image = self.random_hue(image, hue)
        return image
    
    def cutout(self, image, mask_size=16, n_masks=1):
        """
        Cutout: حذف قسمت‌هایی از تصویر
        """
        h, w = tf.shape(image)[0], tf.shape(image)[1]
        
        for _ in range(n_masks):
            # انتخاب موقعیت تصادفی
            y = tf.random.uniform([], 0, h - mask_size, dtype=tf.int32)
            x = tf.random.uniform([], 0, w - mask_size, dtype=tf.int32)
            
            # ایجاد ماسک
            mask = tf.ones([mask_size, mask_size, tf.shape(image)[-1]])
            
            # اعمال ماسک
            image = tf.tensor_scatter_nd_update(
                image,
                tf.stack([
                    tf.range(y, y + mask_size)[:, None, None],
                    tf.range(x, x + mask_size)[None, :, None],
                    tf.range(tf.shape(image)[-1])[None, None, :]
                ], axis=-1),
                tf.zeros([mask_size, mask_size, tf.shape(image)[-1]])
            )
        
        return image
    
    def gaussian_blur(self, image, kernel_size=3, sigma=1.0):
        """
        اعمال فیلتر Gaussian Blur
        """
        # ایجاد کرنل گاوسی
        def gaussian_kernel(size, sigma):
            coords = tf.cast(tf.range(size), tf.float32)
            coords -= (size - 1) / 2.0
            g = tf.exp(-(coords ** 2) / (2 * sigma ** 2))
            g /= tf.reduce_sum(g)
            return g
        
        kernel_1d = gaussian_kernel(kernel_size, sigma)
        kernel_2d = tf.outer(kernel_1d, kernel_1d)
        kernel_3d = tf.expand_dims(tf.expand_dims(kernel_2d, -1), -1)
        
        # اعمال کانولوشن
        image = tf.expand_dims(image, 0)  # اضافه کردن batch dimension
        blurred = tf.nn.conv2d(image, kernel_3d, strides=[1, 1, 1, 1], padding='SAME')
        return tf.squeeze(blurred, 0)  # حذف batch dimension
    
    def elastic_deformation(self, image, alpha=1000, sigma=50):
        """
        تبدیل الاستیک (برای تصاویر پزشکی مفید است)
        """
        # ایجاد فیلد تبدیل تصادفی
        shape = tf.shape(image)
        dx = tf.random.normal([shape[0], shape[1]]) * alpha
        dy = tf.random.normal([shape[0], shape[1]]) * alpha
        
        # صاف کردن فیلد تبدیل
        dx = tf.image.resize(tf.expand_dims(dx, -1), [shape[0], shape[1]])
        dy = tf.image.resize(tf.expand_dims(dy, -1), [shape[0], shape[1]])
        dx = tf.squeeze(dx, -1)
        dy = tf.squeeze(dy, -1)
        
        # ایجاد گرید مختصات
        x, y = tf.meshgrid(tf.range(shape[1]), tf.range(shape[0]))
        x = tf.cast(x, tf.float32) + dx
        y = tf.cast(y, tf.float32) + dy
        
        # نمونه‌برداری مجدد
        coords = tf.stack([y, x], axis=-1)
        return tf.image.resize(image, shape[:2])  # ساده‌سازی شده
    
    def augment_pipeline(self, image, augmentations=None):
        """
        اعمال pipeline کامل augmentation
        """
        if augmentations is None:
            augmentations = {
                'flip': True,
                'rotation': 15,
                'brightness': 0.2,
                'contrast': (0.8, 1.2),
                'cutout': False
            }
        
        # انعکاس تصادفی
        if augmentations.get('flip', False):
            image = self.random_flip(image, 'horizontal')
        
        # چرخش تصادفی
        if 'rotation' in augmentations:
            try:
                image = self.random_rotation(image, augmentations['rotation'])
            except:
                # اگر تابع چرخش در دسترس نباشد، از تبدیل affine استفاده کنیم
                pass
        
        # تغییرات رنگی
        if 'brightness' in augmentations:
            image = self.random_brightness(image, augmentations['brightness'])
        
        if 'contrast' in augmentations:
            contrast = augmentations['contrast']
            image = self.random_contrast(image, contrast[0], contrast[1])
        
        # Cutout
        if augmentations.get('cutout', False):
            image = self.cutout(image)
        
        return image

# تست Data Augmentation
print("=== تست Data Augmentation ===")

# ایجاد تصویر نمونه
test_image = tf.random.uniform((150, 200, 3), 0, 1, dtype=tf.float32)
print(f"تصویر اصلی: {test_image.shape}")

# ایجاد augmentator
augmentator = DataAugmentator()

# تست augmentations مختلف
augmentation_types = [
    ('اصلی', lambda x: x),
    ('انعکاس افقی', lambda x: augmentator.random_flip(x, 'horizontal')),
    ('تغییر روشنایی', lambda x: augmentator.random_brightness(x, 0.3)),
    ('تغییر کنتراست', lambda x: augmentator.random_contrast(x, 0.7, 1.3)),
    ('Color Jitter', lambda x: augmentator.color_jitter(x)),
    ('Pipeline کامل', lambda x: augmentator.augment_pipeline(x))
]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, (name, aug_func) in enumerate(augmentation_types):
    try:
        augmented = aug_func(test_image)
        # اطمینان از محدوده صحیح برای نمایش
        display_img = tf.clip_by_value(augmented, 0, 1)
        axes[i].imshow(display_img.numpy())
        axes[i].set_title(f'{name}\nشکل: {augmented.shape}')
        axes[i].axis('off')
        print(f"✅ {name}: موفق")
    except Exception as e:
        print(f"❌ خطا در {name}: {e}")
        axes[i].text(0.5, 0.5, f'خطا در {name}', ha='center', va='center', 
                    transform=axes[i].transAxes)
        axes[i].axis('off')

plt.tight_layout()
plt.show()

print("\n✅ بخش Data Augmentation تکمیل شد")
```

## ۳.۴ ساخت Dataset و DataLoader

در TensorFlow، tf.data API ابزار قدرتمندی برای مدیریت و پردازش داده‌ها ارائه می‌دهد که امکان ایجاد pipeline‌های کارآمد و مقیاس‌پذیر را فراهم می‌کند.

### مزایای tf.data:
- **کارایی بالا**: پردازش موازی و prefetching خودکار
- **انعطاف‌پذیری**: قابلیت ترکیب transformations مختلف
- **مدیریت حافظه**: بارگذاری تدریجی داده‌ها
- **پشتیبانی از فرمت‌های مختلف**: تصاویر، CSV، TFRecord و غیره
- **Integration با Keras**: سازگاری کامل با model.fit()

### انواع Dataset:
1. **Dataset.from_tensor_slices()**: برای داده‌های در حافظه
2. **Dataset.from_generator()**: برای داده‌های پیچیده
3. **Dataset.list_files()**: برای فایل‌های روی دیسک
4. **image_dataset_from_directory()**: برای مجموعه داده‌های تصویری
5. **TFRecordDataset**: برای فایل‌های TFRecord

### عملیات کلیدی:
- **map()**: اعمال تابع بر روی هر عنصر
- **batch()**: تشکیل batch از داده‌ها
- **shuffle()**: مخلوط کردن داده‌ها
- **repeat()**: تکرار dataset
- **prefetch()**: بارگذاری پیش‌فعال
- **cache()**: ذخیره‌سازی در حافظه یا دیسک

```python
# کلاس مدیریت Dataset برای تصاویر
class ImageDatasetBuilder:
    """
    کلاس جامع برای ساخت و مدیریت Dataset های تصویری
    """
    
    def __init__(self, image_size=(224, 224), batch_size=32, shuffle_buffer=1000):
        self.image_size = image_size
        self.batch_size = batch_size
        self.shuffle_buffer = shuffle_buffer
        self.preprocessor = ImagePreprocessor(image_size)
        self.augmentator = DataAugmentator()
    
    def create_dataset_from_directory(self, directory, validation_split=0.2, 
                                    class_names=None, shuffle=True):
        """
        ایجاد dataset از ساختار فولدری
        
        Structure:
        directory/
            class1/
                img1.jpg
                img2.jpg
            class2/
                img1.jpg
                img2.jpg
        """
        # استفاده از Keras utility
        train_ds = tf.keras.utils.image_dataset_from_directory(
            directory,
            validation_split=validation_split,
            subset="training",
            seed=123,
            image_size=self.image_size,
            batch_size=self.batch_size,
            shuffle=shuffle
        )
        
        val_ds = tf.keras.utils.image_dataset_from_directory(
            directory,
            validation_split=validation_split,
            subset="validation",
            seed=123,
            image_size=self.image_size,
            batch_size=self.batch_size,
            shuffle=shuffle
        )
        
        return train_ds, val_ds
    
    def create_dataset_from_paths_labels(self, image_paths, labels, shuffle=True):
        """
        ایجاد dataset از لیست مسیرها و برچسب‌ها
        """
        dataset = tf.data.Dataset.from_tensor_slices((image_paths, labels))
        
        if shuffle:
            dataset = dataset.shuffle(self.shuffle_buffer)
        
        return dataset
    
    def preprocess_function(self, image_path, label, is_training=False):
        """
        تابع پیش‌پردازش برای هر تصویر
        """
        # خواندن تصویر
        image = tf.io.read_file(image_path)
        image = tf.image.decode_image(image, channels=3)
        image = tf.cast(image, tf.float32)
        
        # پیش‌پردازش
        image = self.preprocessor.preprocess(image)
        
        # Data augmentation برای training
        if is_training:
            image = self.augmentator.augment_pipeline(image)
        
        return image, label
    
    def create_training_pipeline(self, dataset, augment=True, cache=True, prefetch=True):
        """
        ایجاد pipeline آموزشی کامل
        """
        # پیش‌پردازش
        dataset = dataset.map(
            lambda x, y: self.preprocess_function(x, y, is_training=augment),
            num_parallel_calls=tf.data.AUTOTUNE
        )
        
        # کش کردن
        if cache:
            dataset = dataset.cache()
        
        # مخلوط کردن
        dataset = dataset.shuffle(self.shuffle_buffer)
        
        # تشکیل batch
        dataset = dataset.batch(self.batch_size)
        
        # Prefetch
        if prefetch:
            dataset = dataset.prefetch(tf.data.AUTOTUNE)
        
        return dataset
    
    def create_validation_pipeline(self, dataset, cache=True, prefetch=True):
        """
        ایجاد pipeline اعتبارسنجی
        """
        # پیش‌پردازش (بدون augmentation)
        dataset = dataset.map(
            lambda x, y: self.preprocess_function(x, y, is_training=False),
            num_parallel_calls=tf.data.AUTOTUNE
        )
        
        # کش کردن
        if cache:
            dataset = dataset.cache()
        
        # تشکیل batch
        dataset = dataset.batch(self.batch_size)
        
        # Prefetch
        if prefetch:
            dataset = dataset.prefetch(tf.data.AUTOTUNE)
        
        return dataset
    
    def create_inference_pipeline(self, dataset):
        """
        ایجاد pipeline برای inference
        """
        dataset = dataset.map(
            lambda x: self.preprocessor.preprocess(x),
            num_parallel_calls=tf.data.AUTOTUNE
        )
        
        dataset = dataset.batch(self.batch_size)
        dataset = dataset.prefetch(tf.data.AUTOTUNE)
        
        return dataset
    
    def analyze_dataset(self, dataset, num_samples=5):
        """
        تجزیه و تحلیل dataset
        """
        print("=== تحلیل Dataset ===")
        
        # نمایش چند نمونه
        samples = list(dataset.take(num_samples))
        
        if len(samples) > 0:
            first_batch = samples[0]
            if isinstance(first_batch, tuple):
                images, labels = first_batch
                print(f"شکل تصاویر: {images.shape}")
                print(f"شکل برچسب‌ها: {labels.shape}")
                print(f"نوع داده تصاویر: {images.dtype}")
                print(f"محدوده تصاویر: [{tf.reduce_min(images):.3f}, {tf.reduce_max(images):.3f}]")
                print(f"برچسب‌های منحصر به فرد: {tf.unique(tf.reshape(labels, [-1]))[0]}")
            else:
                print(f"شکل داده‌ها: {first_batch.shape}")
                print(f"نوع داده: {first_batch.dtype}")
        
        # محاسبه تعداد کل نمونه‌ها (تقریبی)
        total_batches = 0
        for _ in dataset:
            total_batches += 1
            if total_batches > 100:  # محدود کردن برای جلوگیری از طولانی شدن
                break
        
        estimated_samples = total_batches * self.batch_size
        print(f"تعداد batch ها (تا 100): {total_batches}")
        print(f"تعداد تخمینی نمونه‌ها: {estimated_samples}")
        
        return {
            'batch_size': self.batch_size,
            'estimated_batches': total_batches,
            'estimated_samples': estimated_samples
        }

# تست Dataset Builder
print("=== تست Dataset Builder ===")

# ایجاد dataset builder
dataset_builder = ImageDatasetBuilder(image_size=(128, 128), batch_size=8)

# ساخت dataset نمونه با داده‌های ساختگی
print("\n--- ساخت Dataset نمونه ---")

# ایجاد مسیرهای ساختگی
fake_image_paths = [f"image_{i}.jpg" for i in range(50)]
fake_labels = [i % 5 for i in range(50)]  # 5 کلاس

print(f"تعداد تصاویر: {len(fake_image_paths)}")
print(f"تعداد کلاس‌ها: {len(set(fake_labels))}")
print(f"نمونه مسیرها: {fake_image_paths[:3]}")
print(f"نمونه برچسب‌ها: {fake_labels[:10]}")

# ایجاد dataset
dataset = dataset_builder.create_dataset_from_paths_labels(fake_image_paths, fake_labels)
print(f"\nDataset ایجاد شد: {dataset}")

# نمایش ساختار dataset
print("\n--- ساختار Dataset ---")
for element in dataset.take(3):
    print(f"عنصر: {element}")

# تست pipeline operations
print("\n--- تست Pipeline Operations ---")

# ایجاد dataset با تصاویر واقعی (شبیه‌سازی)
def generate_fake_image_data():
    """
    تولید داده‌های تصویری ساختگی برای تست
    """
    while True:
        # تصویر تصادفی
        image = tf.random.uniform((100, 100, 3), 0, 255, dtype=tf.float32)
        label = tf.random.uniform([], 0, 5, dtype=tf.int32)
        yield image, label

# ایجاد dataset از generator
test_dataset = tf.data.Dataset.from_generator(
    generate_fake_image_data,
    output_signature=(
        tf.TensorSpec(shape=(100, 100, 3), dtype=tf.float32),
        tf.TensorSpec(shape=(), dtype=tf.int32)
    )
)

# محدود کردن تعداد
test_dataset = test_dataset.take(20)

print("Dataset تولید شد از generator")

# اعمال pipeline
processed_dataset = test_dataset.map(
    lambda img, lbl: (dataset_builder.preprocessor.preprocess(img), lbl),
    num_parallel_calls=tf.data.AUTOTUNE
)

processed_dataset = processed_dataset.batch(4)
processed_dataset = processed_dataset.prefetch(tf.data.AUTOTUNE)

print("Pipeline اعمال شد")

# تحلیل dataset
analysis = dataset_builder.analyze_dataset(processed_dataset)
print("\n--- تحلیل Dataset ---")
for key, value in analysis.items():
    print(f"{key}: {value}")

# نمایش چند نمونه
print("\n--- نمایش نمونه‌ها ---")
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

sample_batch = next(iter(processed_dataset))
images, labels = sample_batch

for i in range(min(8, len(images))):
    axes[i].imshow(tf.clip_by_value(images[i], 0, 1))
    axes[i].set_title(f'کلاس: {labels[i].numpy()}')
    axes[i].axis('off')

plt.tight_layout()
plt.show()

print("\n✅ بخش Dataset و DataLoader تکمیل شد")
```

## ۴.۱ Multi-Layer Perceptron (MLP) برای تصاویر

MLP ساده‌ترین نوع شبکه عصبی است که از لایه‌های کاملاً متصل (Dense/Fully Connected) تشکیل شده است. اگرچه برای تصاویر بهینه نیست، اما درک آن برای شروع ضروری است.

### مزایا و معایب MLP:

#### مزایا:
- ساده و قابل فهم
- سریع در آموزش
- مناسب برای تصاویر کوچک
- نیاز کمتر به داده

#### معایب:
- عدم درک ساختار فضایی تصویر
- تعداد پارامتر بالا
- حساسیت به موقعیت اشیا
- عدم translation invariance

### کاربردهای MLP در بینایی ماشین:
- طبقه‌بندی تصاویر ساده
- پردازش feature های استخراج شده
- مرحله نهایی در مدل‌های پیچیده
- مسائل regression بر روی تصاویر

```python
# کلاس ساخت مدل‌های MLP برای تصاویر
class MLPImageClassifier:
    """
    کلاس جامع برای ساخت مدل‌های MLP برای طبقه‌بندی تصویر
    """
    
    def __init__(self, input_shape, num_classes, name="mlp_classifier"):
        self.input_shape = input_shape
        self.num_classes = num_classes
        self.name = name
        self.model = None
        self.history = None
    
    def create_simple_mlp(self, hidden_units=[128, 64], dropout_rate=0.2):
        """
        ایجاد MLP ساده
        
        Args:
            hidden_units: لیست تعداد نورون‌های لایه‌های مخفی
            dropout_rate: نرخ dropout برای regularization
        """
        model = tf.keras.Sequential([
            # Flatten کردن تصویر به بردار 1 بعدی
            tf.keras.layers.Flatten(input_shape=self.input_shape),
            
            # لایه‌های مخفی
            tf.keras.layers.Dense(hidden_units[0], activation='relu'),
            tf.keras.layers.Dropout(dropout_rate),
            
            tf.keras.layers.Dense(hidden_units[1], activation='relu'),
            tf.keras.layers.Dropout(dropout_rate),
            
            # لایه خروجی
            tf.keras.layers.Dense(self.num_classes, activation='softmax')
        ], name=self.name)
        
        self.model = model
        return model
    
    def compile_model(self, optimizer='adam', loss='sparse_categorical_crossentropy',
                     metrics=['accuracy'], learning_rate=0.001):
        """
        کامپایل مدل با تنظیمات مختلف
        """
        if self.model is None:
            raise ValueError("ابتدا مدل را ایجاد کنید")
        
        # تنظیم optimizer
        if optimizer == 'adam':
            opt = tf.keras.optimizers.Adam(learning_rate=learning_rate)
        elif optimizer == 'sgd':
            opt = tf.keras.optimizers.SGD(learning_rate=learning_rate, momentum=0.9)
        elif optimizer == 'rmsprop':
            opt = tf.keras.optimizers.RMSprop(learning_rate=learning_rate)
        else:
            opt = optimizer
        
        self.model.compile(
            optimizer=opt,
            loss=loss,
            metrics=metrics
        )
        
        return self.model

# تست مدل‌های MLP
print("=== تست مدل‌های MLP ===")

# تنظیمات
input_shape = (28, 28, 1)  # مثل MNIST
num_classes = 10

# ایجاد classifier
mlp_classifier = MLPImageClassifier(input_shape, num_classes)

print("\n--- MLP ساده ---")
simple_mlp = mlp_classifier.create_simple_mlp()
mlp_classifier.compile_model()
print(f"تعداد پارامترها: {simple_mlp.count_params():,}")
simple_mlp.summary()

print("\n✅ بخش MLP تکمیل شد")
```

## ۴.۲ Convolutional Neural Networks (CNN)

CNN انقلابی در بینایی ماشین محسوب می‌شود و برای پردازش تصاویر بهینه‌سازی شده است. این مدل‌ها با درک ساختار فضایی تصاویر، عملکرد بهتری نسبت به MLP دارند.

### مزایای CNN:
- **حفظ ساختار فضایی**: درک روابط مکانی پیکسل‌ها
- **Translation Invariance**: عدم حساسیت به جابجایی اشیا
- **Parameter Sharing**: کاهش تعداد پارامترها
- **Feature Hierarchy**: یادگیری ویژگی‌های سلسله‌مراتبی
- **Computational Efficiency**: کارایی محاسباتی بالا

### لایه‌های اصلی CNN:

#### ۱. لایه Convolution:
- استخراج ویژگی با استفاده از فیلترها
- حفظ روابط مکانی
- Parameter sharing

#### ۲. لایه Pooling:
- کاهش ابعاد
- افزایش receptive field
- کاهش overfitting

#### ۳. لایه Activation:
- اضافه کردن غیرخطی بودن
- ReLU محبوب‌ترین انتخاب

```python
# کلاس ساخت مدل‌های CNN پیشرفته
class CNNImageClassifier:
    """
    کلاس جامع برای ساخت مدل‌های CNN برای طبقه‌بندی تصویر
    """
    
    def __init__(self, input_shape, num_classes, name="cnn_classifier"):
        self.input_shape = input_shape
        self.num_classes = num_classes
        self.name = name
        self.model = None
    
    def create_simple_cnn(self):
        """
        ایجاد CNN ساده با چند لایه کانولوشن
        """
        model = tf.keras.Sequential([
            # بلوک کانولوشن اول
            tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=self.input_shape),
            tf.keras.layers.MaxPooling2D((2, 2)),
            
            # بلوک کانولوشن دوم
            tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
            tf.keras.layers.MaxPooling2D((2, 2)),
            
            # بلوک کانولوشن سوم
            tf.keras.layers.Conv2D(128, (3, 3), activation='relu'),
            tf.keras.layers.MaxPooling2D((2, 2)),
            
            # بخش طبقه‌بندی
            tf.keras.layers.Flatten(),
            tf.keras.layers.Dense(128, activation='relu'),
            tf.keras.layers.Dropout(0.5),
            tf.keras.layers.Dense(self.num_classes, activation='softmax')
        ], name=self.name)
        
        self.model = model
        return model
    
    def create_modern_cnn(self, use_batch_norm=True, dropout_rate=0.3):
        """
        ایجاد CNN مدرن با Batch Normalization و Dropout
        """
        model = tf.keras.Sequential(name=self.name)
        
        # بلوک اول
        model.add(tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=self.input_shape))
        if use_batch_norm:
            model.add(tf.keras.layers.BatchNormalization())
        model.add(tf.keras.layers.Conv2D(32, (3, 3), activation='relu'))
        model.add(tf.keras.layers.MaxPooling2D((2, 2)))
        model.add(tf.keras.layers.Dropout(dropout_rate))
        
        # بلوک دوم
        model.add(tf.keras.layers.Conv2D(64, (3, 3), activation='relu'))
        if use_batch_norm:
            model.add(tf.keras.layers.BatchNormalization())
        model.add(tf.keras.layers.Conv2D(64, (3, 3), activation='relu'))
        model.add(tf.keras.layers.MaxPooling2D((2, 2)))
        model.add(tf.keras.layers.Dropout(dropout_rate))
        
        # بخش طبقه‌بندی
        model.add(tf.keras.layers.GlobalAveragePooling2D())  # بهتر از Flatten
        model.add(tf.keras.layers.Dense(256, activation='relu'))
        model.add(tf.keras.layers.Dropout(dropout_rate))
        model.add(tf.keras.layers.Dense(self.num_classes, activation='softmax'))
        
        self.model = model
        return model
    
    def compile_model(self, optimizer='adam', loss='sparse_categorical_crossentropy',
                     metrics=['accuracy'], learning_rate=0.001):
        """
        کامپایل مدل
        """
        if self.model is None:
            raise ValueError("ابتدا مدل را ایجاد کنید")
        
        if optimizer == 'adam':
            opt = tf.keras.optimizers.Adam(learning_rate=learning_rate)
        elif optimizer == 'sgd':
            opt = tf.keras.optimizers.SGD(learning_rate=learning_rate, momentum=0.9)
        else:
            opt = optimizer
        
        self.model.compile(
            optimizer=opt,
            loss=loss,
            metrics=metrics
        )
        
        return self.model

# تست مدل‌های CNN
print("=== تست مدل‌های CNN ===")

# تنظیمات برای تصاویر رنگی
input_shape_rgb = (64, 64, 3)
num_classes = 10

print("\n--- CNN ساده ---")
cnn_simple = CNNImageClassifier(input_shape_rgb, num_classes, "simple_cnn")
simple_model = cnn_simple.create_simple_cnn()
cnn_simple.compile_model()
print(f"تعداد پارامترها: {simple_model.count_params():,}")
simple_model.summary()

print("\n--- CNN مدرن ---")
cnn_modern = CNNImageClassifier(input_shape_rgb, num_classes, "modern_cnn")
modern_model = cnn_modern.create_modern_cnn()
cnn_modern.compile_model()
print(f"تعداد پارامترها: {modern_model.count_params():,}")

print("\n✅ بخش CNN تکمیل شد")
```
